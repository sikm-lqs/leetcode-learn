## 1. 二叉树的中序遍历

- **难度**：简单  
- **标签**：树、深度优先搜索、递归

**题目**：给定一个二叉树的根节点 root，返回它的中序遍历。

**示例**：
```
输入：root = [1, null, 2, 3]
输出：[1, 3, 2]

输入：root = []
输出：[]
```

**思路**：中序遍历顺序为「左 → 中 → 右」。递归实现：先递归遍历左子树，再访问当前节点（加入结果），最后递归遍历右子树；节点为空时递归终止。时间 O(n)，空间 O(h)（递归栈，h 为树高）。

**亮点**：中序遍历的递归结构与二叉搜索树（BST）天然契合——BST 的中序序列恰好是升序；「左-中-右」三行递归是所有树题递归写法的最小骨架。


````python
from typing import Optional, List

# Definition for a binary tree node.
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

def build_tree(vals):
    """按 LeetCode 层序格式构造二叉树，None 表示空位"""
    if not vals:
        return None
    from collections import deque
    root = TreeNode(vals[0])
    queue = deque([root])
    i = 1
    while queue and i < len(vals):
        node = queue.popleft()
        if i < len(vals) and vals[i] is not None:
            node.left = TreeNode(vals[i])
            queue.append(node.left)
        i += 1
        if i < len(vals) and vals[i] is not None:
            node.right = TreeNode(vals[i])
            queue.append(node.right)
        i += 1
    return root

def tree_to_list(root):
    """层序遍历序列化（含 None 占位，去掉尾部 None），便于核对结果"""
    from collections import deque
    if root is None:
        return []
    res, queue = [], deque([root])
    while queue:
        node = queue.popleft()
        if node:
            res.append(node.val)
            queue.append(node.left)
            queue.append(node.right)
        else:
            res.append(None)
    while res and res[-1] is None:
        res.pop()
    return res

class Solution:
    def inorderTraversal(self, root: Optional[TreeNode]) -> List[int]:
        # 中序遍历：左 中间 右
        result = []

        def inorder(node):
            # 递归终止条件
            if node is None:
                return

            # 遍历左子树
            inorder(node.left)

            # 访问当前节点
            result.append(node.val)

            # 遍历右子树
            inorder(node.right)

        inorder(root)

        return result


# 测试
sol = Solution()
print(sol.inorderTraversal(build_tree([1, None, 2, 3])))   # [1, 3, 2]
print(sol.inorderTraversal(build_tree([])))                # []
print(sol.inorderTraversal(build_tree([1])))               # [1]
````

## 2. 对称二叉树

- **难度**：简单  
- **标签**：树、深度优先搜索、递归

**题目**：给你一个二叉树的根节点 root，检查它是否轴对称。

**示例**：
```
输入：root = [1, 2, 2, 3, 4, 4, 3]
输出：true

输入：root = [1, 2, 2, null, 3, null, 3]
输出：false
```

**思路**：对称 = 根节点值相等 + 外侧对外侧 + 内侧对内侧。递归函数 check(left, right) 比较一对镜像节点，按四种情况判定：都空 → True；只有一空 → False；值不等 → False；否则递归比较 check(l.left, r.right)（外侧对）与 check(l.right, r.left)（内侧对）。时间 O(n)，空间 O(h)。

**亮点**：把「整棵树对称」拆解为「每一对镜像节点相等」——比较的对象从同一路径的两个节点变成两条路径的对应节点，是"双参数递归"的入门范式。


````python
from typing import Optional, List

# Definition for a binary tree node.
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

def build_tree(vals):
    """按 LeetCode 层序格式构造二叉树，None 表示空位"""
    if not vals:
        return None
    from collections import deque
    root = TreeNode(vals[0])
    queue = deque([root])
    i = 1
    while queue and i < len(vals):
        node = queue.popleft()
        if i < len(vals) and vals[i] is not None:
            node.left = TreeNode(vals[i])
            queue.append(node.left)
        i += 1
        if i < len(vals) and vals[i] is not None:
            node.right = TreeNode(vals[i])
            queue.append(node.right)
        i += 1
    return root

def tree_to_list(root):
    """层序遍历序列化（含 None 占位，去掉尾部 None），便于核对结果"""
    from collections import deque
    if root is None:
        return []
    res, queue = [], deque([root])
    while queue:
        node = queue.popleft()
        if node:
            res.append(node.val)
            queue.append(node.left)
            queue.append(node.right)
        else:
            res.append(None)
    while res and res[-1] is None:
        res.pop()
    return res

class Solution:
    def isSymmetric(self, root: Optional[TreeNode]) -> bool:
        # 对称 = 根节点值相等 + 外侧对外侧 + 内侧对内侧。递归的思想
        def check(left, right):

            # 1.都为空
            if left is None and right is None:
                return True

            # 2. 只有一个为空
            if left is None or right is None:
                return False

            # 3.两个节点的val不同
            if left.val != right.val:
                return False

            # 4. 镜像比较
            return check(left.left, right.right) and check(left.right, right.left)

        return check(root.left, root.right)


# 测试
sol = Solution()
print(sol.isSymmetric(build_tree([1, 2, 2, 3, 4, 4, 3])))          # True
print(sol.isSymmetric(build_tree([1, 2, 2, None, 3, None, 3])))    # False
print(sol.isSymmetric(build_tree([1])))                            # True（单节点）
````

## 3. 二叉树的最大深度

- **难度**：简单  
- **标签**：树、深度优先搜索、递归

**题目**：给定一个二叉树 root，返回其最大深度。二叉树的最大深度是指从根节点到最远叶子节点的最长路径上的节点数。

**示例**：
```
输入：root = [3, 9, 20, null, null, 15, 7]
输出：3

输入：root = [1, null, 2]
输出：2
```

**思路**：递归定义——树的最大深度 = max(左子树深度, 右子树深度) + 1。空树深度为 0（终止条件），当前节点的深度由左右子树答案汇合而来。时间 O(n)，空间 O(h)。

**亮点**：「分解问题」式递归的标准样本——不关心路径细节，只信任子问题的返回值；这一行 return max(left, right) + 1 是上百道树题（直径、平衡、路径和……）的公共内核。


````python
from typing import Optional, List

# Definition for a binary tree node.
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

def build_tree(vals):
    """按 LeetCode 层序格式构造二叉树，None 表示空位"""
    if not vals:
        return None
    from collections import deque
    root = TreeNode(vals[0])
    queue = deque([root])
    i = 1
    while queue and i < len(vals):
        node = queue.popleft()
        if i < len(vals) and vals[i] is not None:
            node.left = TreeNode(vals[i])
            queue.append(node.left)
        i += 1
        if i < len(vals) and vals[i] is not None:
            node.right = TreeNode(vals[i])
            queue.append(node.right)
        i += 1
    return root

def tree_to_list(root):
    """层序遍历序列化（含 None 占位，去掉尾部 None），便于核对结果"""
    from collections import deque
    if root is None:
        return []
    res, queue = [], deque([root])
    while queue:
        node = queue.popleft()
        if node:
            res.append(node.val)
            queue.append(node.left)
            queue.append(node.right)
        else:
            res.append(None)
    while res and res[-1] is None:
        res.pop()
    return res

class Solution:
    def maxDepth(self, root: Optional[TreeNode]) -> int:
        # max(左子树深度, 右子树深度) + 1，递归思想

        # 空树 depth = 0
        if root is None:
            return 0

        # 计算左/右子树的最大深度
        left_depth = self.maxDepth(root.left)
        right_depth = self.maxDepth(root.right)

        # 当前树的最大深度为左右更深的 +1
        return max(left_depth, right_depth) + 1


# 测试
sol = Solution()
print(sol.maxDepth(build_tree([3, 9, 20, None, None, 15, 7])))  # 3
print(sol.maxDepth(build_tree([1, None, 2])))                   # 2
print(sol.maxDepth(build_tree([])))                             # 0
````

## 4. 翻转二叉树

- **难度**：简单  
- **标签**：树、深度优先搜索、递归

**题目**：给你一棵二叉树的根节点 root，翻转这棵二叉树，并返回其根节点。

**示例**：
```
输入：root = [4, 2, 7, 1, 3, 6, 9]
输出：[4, 7, 2, 9, 6, 3, 1]
```

**思路**：递归思想。终止条件：当前节点为空返回 None。单层逻辑：交换当前节点的左右孩子，然后递归翻转左子树和右子树，最后返回根节点。时间 O(n)，空间 O(h)。

**亮点**：三行核心（判空/交换/递归）完成的经典"修改树"题——注意「先交换再递归」与「先递归再交换」都正确，交换的是孩子指针本身，与子树内部翻转顺序无关。


````python
from typing import Optional, List

# Definition for a binary tree node.
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

def build_tree(vals):
    """按 LeetCode 层序格式构造二叉树，None 表示空位"""
    if not vals:
        return None
    from collections import deque
    root = TreeNode(vals[0])
    queue = deque([root])
    i = 1
    while queue and i < len(vals):
        node = queue.popleft()
        if i < len(vals) and vals[i] is not None:
            node.left = TreeNode(vals[i])
            queue.append(node.left)
        i += 1
        if i < len(vals) and vals[i] is not None:
            node.right = TreeNode(vals[i])
            queue.append(node.right)
        i += 1
    return root

def tree_to_list(root):
    """层序遍历序列化（含 None 占位，去掉尾部 None），便于核对结果"""
    from collections import deque
    if root is None:
        return []
    res, queue = [], deque([root])
    while queue:
        node = queue.popleft()
        if node:
            res.append(node.val)
            queue.append(node.left)
            queue.append(node.right)
        else:
            res.append(None)
    while res and res[-1] is None:
        res.pop()
    return res

class Solution:
    def invertTree(self, root: Optional[TreeNode]) -> Optional[TreeNode]:
        # 递归思想

        # 1.递归终止条件：当前节点为空
        if root is None:
            return None

        # 2. 交换
        root.left, root.right = root.right, root.left

        # 递归
        self.invertTree(root.left)
        self.invertTree(root.right)

        return root


# 测试
sol = Solution()
root = sol.invertTree(build_tree([4, 2, 7, 1, 3, 6, 9]))
print(tree_to_list(root))  # [4, 7, 2, 9, 6, 3, 1]
print(tree_to_list(sol.invertTree(build_tree([2, 1, 3]))))  # [2, 3, 1]
print(tree_to_list(sol.invertTree(build_tree([]))))         # []
````

## 5. 二叉树的直径

- **难度**：简单  
- **标签**：树、深度优先搜索、递归

**题目**：给你一棵二叉树的根节点，返回该树的直径。二叉树的直径是指树中任意两个节点之间最长路径的长度，这条路径可能经过也可能不经过根节点。两节点之间路径的长度由它们之间的边数表示。

**示例**：
```
输入：root = [1, 2, 3, 4, 5]
输出：3
解释：最长路径为 4 → 2 → 1 → 3 或 5 → 2 → 1 → 3，长度 3（边数）。

输入：root = [1, 2]
输出：1
```

**思路**：经过当前节点的最长路径（边数）= 左子树深度 + 右子树深度。在计算每个节点深度的递归中，顺便用 nonlocal ans 记录 max(ans, left + right)，递归返回的仍是该子树深度 max(left, right) + 1。时间 O(n)，空间 O(h)。

**亮点**：「一次遍历两个答案」——递归函数的返回值是深度（给父节点用），副作用是更新直径（全局答案）；返回值与统计量分离，是"在遍历中求最值"类题目的通用手法（最大路径和同构）。


````python
from typing import Optional, List

# Definition for a binary tree node.
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

def build_tree(vals):
    """按 LeetCode 层序格式构造二叉树，None 表示空位"""
    if not vals:
        return None
    from collections import deque
    root = TreeNode(vals[0])
    queue = deque([root])
    i = 1
    while queue and i < len(vals):
        node = queue.popleft()
        if i < len(vals) and vals[i] is not None:
            node.left = TreeNode(vals[i])
            queue.append(node.left)
        i += 1
        if i < len(vals) and vals[i] is not None:
            node.right = TreeNode(vals[i])
            queue.append(node.right)
        i += 1
    return root

def tree_to_list(root):
    """层序遍历序列化（含 None 占位，去掉尾部 None），便于核对结果"""
    from collections import deque
    if root is None:
        return []
    res, queue = [], deque([root])
    while queue:
        node = queue.popleft()
        if node:
            res.append(node.val)
            queue.append(node.left)
            queue.append(node.right)
        else:
            res.append(None)
    while res and res[-1] is None:
        res.pop()
    return res

class Solution:
    def diameterOfBinaryTree(self, root: Optional[TreeNode]) -> int:
        # 每个节点的直径等于左右深度相加：整棵树中，两个节点之间最长路径的边数，
        # 经过当前节点的直径 = 左子树深度 + 右子树深度
        # 对每一个节点，都计算一次 left + right，然后取最大的。

        ans = 0

        def depth(node):
            nonlocal ans  # 声明非局部

            if not node:
                return 0

            left = depth(node.left)
            right = depth(node.right)

            # 更新 ans
            ans = max(ans, left + right)

            return max(left, right) + 1

        depth(root)
        return ans


# 测试
sol = Solution()
print(sol.diameterOfBinaryTree(build_tree([1, 2, 3, 4, 5])))  # 3
print(sol.diameterOfBinaryTree(build_tree([1, 2])))           # 1
print(sol.diameterOfBinaryTree(build_tree([1])))              # 0
````

## 6. 合并二叉树

- **难度**：简单  
- **标签**：树、深度优先搜索、递归

**题目**：给你两棵二叉树 root1 和 root2。将其中一棵覆盖到另一棵之上时，一些节点会重叠。合并规则：如果两个节点重叠，将这两个节点的值相加作为合并后节点的新值；否则不为 null 的节点将直接作为新二叉树的节点。返回合并后的二叉树（合并过程必须从两个树的根节点开始）。

**示例**：
```
输入：root1 = [1, 3, 2, 5]，root2 = [2, 1, 3, null, 4, null, 7]
输出：[3, 4, 5, 5, 4, null, 7]

输入：root1 = [1]，root2 = [1, 2]
输出：[2, 2]
```

**思路**：递归思想——同时看两棵树相同位置的节点：一棵为空直接返回另一棵；都存在则值相加（本题在 root1 上原地合并），再递归合并左右孩子并接回。时间 O(min(m, n))（较短树先耗尽），空间 O(min(m, n))。

**亮点**：「双树同走」递归——参数是两个节点同步推进；缺位子树整棵挂接（返回另一棵），不再深入递归，天然剪枝。


````python
from typing import Optional, List

# Definition for a binary tree node.
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

def build_tree(vals):
    """按 LeetCode 层序格式构造二叉树，None 表示空位"""
    if not vals:
        return None
    from collections import deque
    root = TreeNode(vals[0])
    queue = deque([root])
    i = 1
    while queue and i < len(vals):
        node = queue.popleft()
        if i < len(vals) and vals[i] is not None:
            node.left = TreeNode(vals[i])
            queue.append(node.left)
        i += 1
        if i < len(vals) and vals[i] is not None:
            node.right = TreeNode(vals[i])
            queue.append(node.right)
        i += 1
    return root

def tree_to_list(root):
    """层序遍历序列化（含 None 占位，去掉尾部 None），便于核对结果"""
    from collections import deque
    if root is None:
        return []
    res, queue = [], deque([root])
    while queue:
        node = queue.popleft()
        if node:
            res.append(node.val)
            queue.append(node.left)
            queue.append(node.right)
        else:
            res.append(None)
    while res and res[-1] is None:
        res.pop()
    return res

class Solution:
    def mergeTrees(self, root1: Optional[TreeNode], root2: Optional[TreeNode]) -> Optional[TreeNode]:
        # 递归思想：同时看两棵树当前位置的节点：
        # 都有就相加，只有一个就用那个，然后继续处理左右孩子

        # 1. 只有一棵树存在，直接返回另一棵
        if root1 is None:
            return root2

        if root2 is None:
            return root1

        # 2. 两个节点都存在，值相加
        root1.val += root2.val

        # 3. 继续合并左子树和右子树
        root1.left = self.mergeTrees(root1.left, root2.left)
        root1.right = self.mergeTrees(root1.right, root2.right)

        return root1


# 测试
sol = Solution()
merged = sol.mergeTrees(
    build_tree([1, 3, 2, 5]),
    build_tree([2, 1, 3, None, 4, None, 7]),
)
print(tree_to_list(merged))  # [3, 4, 5, 5, 4, None, 7]
print(tree_to_list(sol.mergeTrees(build_tree([1]), build_tree([1, 2]))))  # [2, 2]
````

## 7. 不同的二叉搜索树

- **难度**：中等  
- **标签**：树、动态规划、卡特兰数

**题目**：给你一个整数 n，求恰由 n 个节点组成且节点值从 1 到 n 互不相同的**二叉搜索树**有多少种？返回满足题意的二叉搜索树的种数。

**示例**：
```
输入：n = 3
输出：5
解释：节点 1、2、3 可组成 5 种不同结构的 BST（根为 1/2/3 时左右子树规模的划分各不相同）。

输入：n = 1
输出：1
```

**思路**：枚举根节点。值 root 做根时，左子树必须由 1..root-1（共 root-1 个节点）组成，右子树由 root+1..n（共 nodes-root 个节点）组成。**BST 的结构种数只与节点个数有关，与具体值无关**，所以 dp[k] = k 个节点能组成的 BST 种数，转移为左右子树种数的乘积再对所有根求和：dp[nodes] = Σ dp[root-1] × dp[nodes-root]（root = 1..nodes），dp[0] = 1（空树一种）。这就是卡特兰数递推。DP 版 O(n²) 时间 / O(n) 空间；递归版用 @cache 记忆化同一转移式。

**亮点**：两个关键洞察——①「枚举根」把整棵树的构造拆成独立并列的左右子树（乘法原理）；②「种数只依赖节点个数」让值 1..n 与任意 k 个有序数等价，状态从「值的集合」压缩成一个整数 k。


````python
from functools import cache

class Solution:
    def numTrees(self, n: int) -> int:
        # dp[k]：用 k 个节点，可以组成多少种不同的二叉搜索树
        # 二叉搜索树：左子树所有值 < 根节点 < 右子树所有值
        # dp[节点总数] += 左子树种数 * 右子树种数
        # dp[n] = dp[0]*dp[n-1] + dp[1]*dp[n-2]...

        dp = [0] * (n + 1)

        dp[0] = 1

        for nodes in range(1, n + 1):
            for root in range(1, nodes + 1):
                left = root - 1
                right = nodes - root

                dp[nodes] += dp[left] * dp[right]

        return dp[n]


class Solution2:
    def numTrees(self, n: int) -> int:
        # 递归版本
        @cache
        def dfs(n):
            # 0 或者 1 个节点，都只有一种情况
            if n <= 1:
                return 1

            ans = 0

            # 枚举根节点
            for root in range(1, n + 1):
                left = root - 1
                right = n - root

                ans += dfs(left) * dfs(right)

            return ans

        return dfs(n)


# 测试
sol = Solution()
print(sol.numTrees(3))    # 5
print(sol.numTrees(1))    # 1
print(sol.numTrees(4))    # 14
print(sol.numTrees(19))   # 1767263190

sol2 = Solution2()
print(sol2.numTrees(3))   # 5
print(sol2.numTrees(19))  # 1767263190
````

## 8. 验证二叉搜索树

- **难度**：中等  
- **标签**：树、深度优先搜索、二叉搜索树

**题目**：给你一个二叉树的根节点 root，判断其是否是一个有效的二叉搜索树。有效二叉搜索树定义如下：
- 节点的左子树只包含**严格小于**当前节点的数
- 节点的右子树只包含**严格大于**当前节点的数
- 所有左子树和右子树自身必须也是二叉搜索树

**示例**：
```
输入：root = [2, 1, 3]
输出：true

输入：root = [5, 1, 4, null, null, 3, 6]
输出：false
解释：根的右子树包含 3 和 6 < 5，违反"右子树严格大于根"。

输入：root = [5, 4, 6, null, null, 3, 7]
输出：false
解释：每个节点与直接父节点比较都合法，但节点 3 < 5 违反祖先约束——只看父子会漏判。
```

**思路**：递归函数 dfs(node, lower, upper) 携带开区间 (lower, upper) 下降。当前节点必须严格落在区间内；进入左子树时上界收紧为 node.val（左子树全体必须 < node.val），进入右子树时下界收紧为 node.val。初始区间 (-∞, +∞)。区间把"全部祖先的约束"压缩成两个参数沿路径传递。时间 O(n)（每个节点访问一次），空间 O(h)（递归栈）。

**亮点**：BST 的合法性不是局部性质——节点只与父比较不够（反例 [5,4,6,null,null,3,7]）；上下界法把祖先链的所有约束浓缩为 (lower, upper) 一对参数，一次后序遍历完成验证。等价解法（Solution2）：BST 的中序遍历恰好严格递增——中序收集到数组后检查相邻 nums[i] > nums[i-1] 即可，O(n) 时间 / O(n) 空间。


````python
from typing import Optional, List

# Definition for a binary tree node.
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

def build_tree(vals):
    """按 LeetCode 层序格式构造二叉树，None 表示空位"""
    if not vals:
        return None
    from collections import deque
    root = TreeNode(vals[0])
    queue = deque([root])
    i = 1
    while queue and i < len(vals):
        node = queue.popleft()
        if i < len(vals) and vals[i] is not None:
            node.left = TreeNode(vals[i])
            queue.append(node.left)
        i += 1
        if i < len(vals) and vals[i] is not None:
            node.right = TreeNode(vals[i])
            queue.append(node.right)
        i += 1
    return root

class Solution:
    def isValidBST(self, root: Optional[TreeNode]) -> bool:
        # 定义一个递归函数，从 root 开始，判断 root.left < root < root.right，
        # 然后继续把 root.left / root.right 当做 root，直到某个 root.left/right == None
        # 有个问题是不能只考虑父亲节点，还要考虑祖先节点
        def dfs(root, lower, upper):
            if root is None:
                return True

            # 当前节点必须要在 lower-upper 之间
            if root.val <= lower or root.val >= upper:
                return False

            # 左子树（上界收紧为 root.val）
            left = dfs(root.left, lower, root.val)
            right = dfs(root.right, root.val, upper)

            return left and right

        return dfs(root, float("-inf"), float("inf"))


class Solution2:
    def isValidBST(self, root: Optional[TreeNode]) -> bool:
        # 中序遍历：左子树 -> root -> 右子树
        # 将中序遍历保存到数组，然后检查

        nums = []

        # 中序遍历
        def inorder(node):
            if not node:
                return
            # 1.遍历左子树
            inorder(node.left)

            # 2. 当前值
            nums.append(node.val)

            # 3. 右子树
            inorder(node.right)

        inorder(root)

        for i in range(1, len(nums)):
            if nums[i] <= nums[i-1]:
                return False

        return True


# 测试
sol = Solution()
print(sol.isValidBST(build_tree([2, 1, 3])))                      # True
print(sol.isValidBST(build_tree([5, 1, 4, None, None, 3, 6])))    # False
print(sol.isValidBST(build_tree([5, 4, 6, None, None, 3, 7])))    # False（祖先约束）
print(sol.isValidBST(build_tree([])))                             # True（空树）

# 中序遍历解法对照
sol2 = Solution2()
print(sol2.isValidBST(build_tree([2, 1, 3])))                      # True
print(sol2.isValidBST(build_tree([5, 1, 4, None, None, 3, 6])))    # False
print(sol2.isValidBST(build_tree([5, 4, 6, None, None, 3, 7])))    # False
print(sol2.isValidBST(build_tree([])))                             # True
````

## 9. 二叉树的层序遍历

- **难度**：中等  
- **标签**：树、广度优先搜索（BFS）、队列

**题目**：给你二叉树的根节点 root，返回其节点值的**层序遍历**（即逐层地，从左到右访问所有节点）。

**示例**：
```
输入：root = [3, 9, 20, null, null, 15, 7]
        3
       / \
      9  20
         / \
        15  7
输出：[[3], [9, 20], [15, 7]]

输入：root = [1]
输出：[[1]]

输入：root = []
输出：[]
```

**思路**：BFS 分层模板。根节点入队；while 队列非空时，先记录当前层的节点数 len(queue)，用 for 循环一次性把这层全部取出（记录值、左右孩子依次入队），取出的值组成 level 加入结果。孩子在入队前判空可省去 None 检查。时间 O(n)（每个节点入队出队各一次），空间 O(w)（w 为最宽层的节点数）。

**亮点**：「len(queue) 定格层大小」是分层的灵魂——循环开始那一刻队列里恰好是完整的一层，取出过程中新入队的孩子属于下一层、不会混入本轮；配合 deque 的 popleft() O(1) 出队，整套模板可直接迁移到"每层最右值""锯齿遍历""最小深度"等一整族题。


````python
from collections import deque
from typing import Optional, List

# Definition for a binary tree node.
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

def build_tree(vals):
    """按 LeetCode 层序格式构造二叉树，None 表示空位"""
    if not vals:
        return None
    from collections import deque
    root = TreeNode(vals[0])
    queue = deque([root])
    i = 1
    while queue and i < len(vals):
        node = queue.popleft()
        if i < len(vals) and vals[i] is not None:
            node.left = TreeNode(vals[i])
            queue.append(node.left)
        i += 1
        if i < len(vals) and vals[i] is not None:
            node.right = TreeNode(vals[i])
            queue.append(node.right)
        i += 1
    return root

class Solution:
    def levelOrder(self, root: Optional[TreeNode]) -> List[List[int]]:
        # 层序遍历
        # 1. root 放进队列
        # 2. while 队列不为空：创建 level
        # 3. 把当前这一层全部取出来：记录 node.val、左孩子入队、右孩子入队
        # 4. level 加入 res

        if not root:
            return []

        queue = deque([root])
        res = []

        while queue:
            level = []

            # 当前层有多少节点（循环开始时定格，新入队的孩子不影响本轮）
            for _ in range(len(queue)):
                node = queue.popleft()
                level.append(node.val)

                if node.left:
                    queue.append(node.left)
                if node.right:
                    queue.append(node.right)

            res.append(level)

        return res


# 测试
sol = Solution()
print(sol.levelOrder(build_tree([3, 9, 20, None, None, 15, 7])))  # [[3], [9, 20], [15, 7]]
print(sol.levelOrder(build_tree([1])))                            # [[1]]
print(sol.levelOrder(build_tree([])))                             # []
````
