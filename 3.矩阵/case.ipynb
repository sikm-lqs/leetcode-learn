```python
# 统计数字出现次数
nums = [1,2,2,3,3,3]

cnt = {}

for x in nums:
    cnt[x] = cnt.get(x, 0) + 1

print(cnt)
```

```python
# 滑动窗口
nums = [2, 3, 1, 2, 4, 3]
target = 7

left = 0
window_sum = 0
ans = float('inf')

for right in range(len(nums)):
    window_sum += nums[right]

    while window_sum >= target:
        ans = min(ans, right -left + 1)
        window_sum -= nums[left]
        left += 1

print(ans)
```
