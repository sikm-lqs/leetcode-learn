## 1. 两数之和

- **难度**：简单  
- **标签**：哈希表

**题目**：给定整数数组 `nums` 和目标值 `target`，找出和为目标值的两个整数，返回它们的下标。

**示例**：`nums = [2, 7, 11, 15], target = 9` → 输出 `[0, 1]`

**思路**：遍历时用哈希表记录已见过的数及其下标。对于每个数 `num`，检查 `target - num` 是否已在表中——是则直接返回下标对；否则将当前数存入表中。时间 O(n)，空间 O(n)。

**亮点**：边遍历边记录，第一次遍历即可找到答案，无需两次遍历。

````python
from typing import List

class Solution:
    def twoSum(self, nums: List[int], target: int) -> List[int]:
        num_map = {}
        for idx, num in enumerate(nums):
            complement = target - num
            if complement in num_map:
                return [num_map[complement], idx]
            num_map[num] = idx

# 测试
sol = Solution()
print(sol.twoSum([2, 7, 11, 15], 9))  # [0, 1]
````

## 2. 移动零

- **难度**：简单  
- **标签**：双指针

**题目**：将数组中所有 0 移动到末尾，同时保持非零元素的相对顺序。原地操作。

**示例**：`nums = [0, 1, 0, 3, 12]` → 输出 `[1, 3, 12, 0, 0]`

**思路**：快慢指针。快指针 `fast` 遍历数组找非零元素，慢指针 `slow` 指向下一个非零元素应放入的位置。遇到非零即交换，交换后 `slow` 前移。时间 O(n)，空间 O(1)。

**亮点**：一趟遍历完成，且交换操作同时将零移至末尾。

````python
from typing import List

class Solution:
    def moveZeroes(self, nums: List[int]) -> None:
        slow = 0
        for fast in range(len(nums)):
            if nums[fast] != 0:
                nums[slow], nums[fast] = nums[fast], nums[slow]
                slow += 1

# 测试
nums = [0, 1, 0, 3, 12]
Solution().moveZeroes(nums)
print(nums)  # [1, 3, 12, 0, 0]
````

## 3. 盛最多水的容器

- **难度**：中等  
- **标签**：双指针

**题目**：给定高度数组，找出两条线使它们与 x 轴构成的容器能容纳最多的水。

**示例**：`height = [1, 8, 6, 2, 5, 4, 8, 3, 7]` → 输出 `49`

**思路**：双指针从两端向中间收缩。面积由 `min(height[left], height[right]) * (right - left)` 决定。已知宽度固定，要增大面积，必须移动较短边的指针，因为移动较长边只会让高度不变或减小而宽度减小。时间 O(n)，空间 O(1)。

**亮点**：左右夹逼，每一次移动都是当前局部的最优决策，最终必得全局最优。

````python
from typing import List

class Solution:
    def maxArea(self, height: List[int]) -> int:
        left, right = 0, len(height) - 1
        max_area = 0
        while left < right:
            cur_area = (right - left) * min(height[left], height[right])
            max_area = max(max_area, cur_area)
            if height[left] < height[right]:
                left += 1
            else:
                right -= 1
        return max_area

# 测试
sol = Solution()
print(sol.maxArea([1, 8, 6, 2, 5, 4, 8, 3, 7]))  # 49
````

## 4. 三数之和

- **难度**：中等  
- **标签**：双指针、排序

**题目**：返回数组中和为 0 的不重复三元组。

**示例**：`nums = [-1, 0, 1, 2, -1, -4]` → 输出 `[[-1, -1, 2], [-1, 0, 1]]`

**思路**：排序 + 双指针。先将数组排序，固定一个数 `i`，然后用双指针 `left`、`right` 在 `i+1` 和 `n-1` 之间寻找两数之和为 `-nums[i]`。total < 0 则 left 右移，total > 0 则 right 左移，等于 0 则加入结果并同时收缩左右指针去重。时间 O(n²)，空间 O(1)。

**亮点**：排序后可用双指针将两数和问题从 O(n) 降至 O(1)，整体 O(n²) 而非暴力 O(n³)；去重逻辑细腻，三处去重分别处理首位重复和答案重复。

````python
class Solution:
    def threeSum(self, nums: list[int]) -> list[list[int]]:
        nums.sort()
        res = []
        n = len(nums)
        for i in range(n):
            if i > 0 and nums[i] == nums[i - 1]:
                continue
            left, right = i + 1, n - 1
            while left < right:
                total = nums[i] + nums[left] + nums[right]
                if total == 0:
                    res.append([nums[i], nums[left], nums[right]])
                    while left < right and nums[left] == nums[left + 1]:
                        left += 1
                    while left < right and nums[right] == nums[right - 1]:
                        right -= 1
                    left += 1
                    right -= 1
                elif total < 0:
                    left += 1
                else:
                    right -= 1
        return res

# 测试
sol = Solution()
print(sol.threeSum([-1, 0, 1, 2, -1, -4]))  # [[-1, -1, 2], [-1, 0, 1]]
````

## 5. 合并区间

- **难度**：中等  
- **标签**：排序、双指针

**题目**：给定若干区间，合并所有重叠的区间，返回恰好覆盖输入中所有区间的不重叠区间数组。

**示例**：`intervals = [[1,3],[2,6],[8,10],[15,18]]` → 输出 `[[1,6],[8,10],[15,18]]`

**思路**：先按起点排序。遍历时维护结果数组 `res`，若当前区间起点 > `res` 最后一个区间的终点，则无重叠，直接加入；否则有重叠，更新 `res` 最后一个区间的终点为两者较大值。时间 O(n log n)，空间 O(1)。

**亮点**：排序后只需 O(n) 一次遍历，重叠判断只需比较左端点与已合并区间的右端点，无需复杂数据结构。

````python
from typing import List

class Solution:
    def merge(self, intervals: List[List[int]]) -> List[List[int]]:
        intervals.sort()
        res = []
        for cur in intervals:
            if not res or cur[0] > res[-1][1]:
                res.append(cur)
            else:
                res[-1][1] = max(res[-1][1], cur[1])
        return res

# 测试
sol = Solution()
print(sol.merge([[1,3],[2,6],[8,10],[15,18]]))  # [[1,6],[8,10],[15,18]]
````

## 6. 轮转数组

- **难度**：中等  
- **标签**：双指针、数组

**题目**：给定整数数组 `nums`，将其元素向右轮转 `k` 个位置，`k` 为非负数。原地修改。

**示例**：`nums = [1,2,3,4,5,6,7], k = 3` → 输出 `[5,6,7,1,2,3,4]`

**思路**：三次反转。先将 `k` 取余避免越界，整体反转，再分别反转前 `k` 个和后 `n-k` 个。最终效果即为整体右旋 `k` 位。时间 O(n)，空间 O(1)。

**亮点**：利用反转的叠加性实现整体轮转，单一技巧解决，无需额外数组。

````python
from typing import List

class Solution:
    def rotate(self, nums: List[int], k: int) -> None:
        k = k % len(nums)
        nums.reverse()
        self._reverse(0, k - 1, nums)   
        self._reverse(k, len(nums) - 1, nums)

    def _reverse(self, start: int, end: int, num: List[int]) -> None:
        while start < end:
            num[start], num[end] = num[end], num[start]
            start += 1
            end -= 1

# 测试
nums = [1, 2, 3, 4, 5, 6, 7]
Solution().rotate(nums, 3)
print(nums)  # [5, 6, 7, 1, 2, 3, 4]
````

## 7. 除自身以外数组的乘积

- **难度**：中等
- **标签**：数组、前缀和

**题目**：给定整数数组 `nums`，返回数组 `answer`，其中 `answer[i]` 等于 `nums` 中除 `nums[i]` 之外其余各元素的乘积。不使用除法，且在 O(n) 时间复杂度内完成。

**示例**：
```python
输入：nums = [1, 2, 3, 4]
输出：[24, 12, 8, 6]
解释：
  answer[0] = 2 * 3 * 4 = 24
  answer[1] = 1 * 3 * 4 = 12
  answer[2] = 1 * 2 * 4 = 8
  answer[3] = 1 * 2 * 3 = 6
```

**思路**：
1. **第一次遍历（从左到右）**：计算每个位置左侧所有数字的乘积，存入 `answer[i]`
2. **第二次遍历（从右到左）**：用一个变量 `right_product` 记录右侧乘积，乘到 `answer[i]` 上
   - 时间复杂度：O(n)
   - 空间复杂度：O(1)，输出数组不计入空间复杂度

**亮点**：两次遍历，分别计算前缀积和后缀积，用输出数组暂存前缀积，避免了额外的空间分配；不使用除法，完美处理了数组中有 0 的情况。

````python
from typing import List

class Solution:
    def productExceptSelf(self, nums: list[int]) -> list[int]:
        n = len(nums)
        answer = [1] * n

        # 第一遍：从左往右
        # answer[i] 表示 nums[i] 左边所有元素的乘积
        #
        # 例如 nums = [1, 2, 3, 4]
        # 最终得到：
        # answer = [1, 1, 2, 6]
        for i in range(1, n):
            answer[i] = answer[i - 1] * nums[i - 1]

        # right 表示当前位置右边所有元素的乘积
        right = 1

        # 第二遍：从右往左
        # 把右侧乘积乘到 answer[i] 上
        for i in range(n - 1, -1, -1):
            answer[i] *= right

            # 更新 right，供下一个位置使用
            right *= nums[i]

        return answer

# 测试
sol = Solution()
print(sol.productExceptSelf([1, 2, 3, 4]))  # [24, 12, 8, 6]
print(sol.productExceptSelf([-1, 1, 0, -3, 3]))  # [0, 0, 9, 0, 0]
````

## 8. 和为 K 的子数组

- **难度**：中等
- **标签**：数组、哈希表、前缀和

**题目**：给定整数数组 `nums` 和整数 `k`，统计并返回该数组中和为 `k` 的子数组个数。子数组是数组中元素的连续非空序列。

**示例**：
```python
输入：nums = [1, 1, 1], k = 2
输出：2
解释：有两个连续子数组和为 2：[1, 1] (索引 0-1) 和 [1, 1] (索引 1-2)

输入：nums = [1, 2, 3], k = 3
输出：2
解释：有两个连续子数组和为 3：[1, 2] 和 [3]
```

**思路**：前缀和 + 哈希表。定义 `pre[i]` 为 `nums[0] + nums[1] + ... + nums[i]`。若 `nums[l→r]` 和为 `k`，则 `pre[r] - pre[l-1] = k`，即 `pre[l-1] = pre[r] - k`。遍历时用哈希表记录各前缀和出现的次数，对于当前前缀和 `s`，查找表中 `s - k` 的次数并累加到答案。时间 O(n)，空间 O(n)。

**亮点**：将子数组求和转化为「前缀和差值」问题，用哈希表存储前缀和频率，将 O(n²) 暴力枚举降至 O(n) 单次遍历；初始化 `count[0] = 1` 处理了从索引 0 开始的子数组。

````python
from typing import List

class Solution:
    def subarraySum(self, nums: List[int], k: int) -> int:
        # 前缀和 + 哈希表
        # pre[i] = nums[0] + nums[1] + ... + nums[i]
        # 求 nums[l→r] = pre[r] - pre[l-1]
        
        # 哈希表：key 是前缀和，value 是出现次数
        count = {0: 1}  # 初始化：前缀和为 0 出现过 1 次（处理从索引 0 开始的子数组）
        
        s = 0  # 当前前缀和
        ans = 0

        for num in nums:
            s += num
            # 若存在前缀和为 s - k，则这些位置到当前位置的子数组和为 k
            ans += count.get(s - k, 0)
            # 记录当前前缀和的出现次数
            count[s] = count.get(s, 0) + 1
        
        return ans

# 测试
sol = Solution()
print(sol.subarraySum([1, 1, 1], 2))  # 2
print(sol.subarraySum([1, 2, 3], 3))  # 2
print(sol.subarraySum([1, -1, 0], 0))  # 3: [1,-1], [-1,0], [0]
print(sol.subarraySum([10, 2, -2, -20, 10], -10))  # 3: [10,2,-2,-20], [2,-2,-20,10], [-20,10]
````

## 9. 最长连续序列

- **难度**：中等
- **标签**：数组、哈希表、集合

**题目**：给定一个未排序的整数数组 `nums`，找出数字连续的最长序列的长度。要求时间复杂度为 O(n)（不要求序列元素在原数组中连续）。

**示例**：
```python
输入：nums = [100, 4, 200, 1, 3, 2]
输出：4
解释：最长数字连续序列是 [1, 2, 3, 4]，长度为 4

输入：nums = [0, 3, 7, 2, 5, 8, 4, 6, 0, 1]
输出：9
解释：最长数字连续序列是 [0, 1, 2, 3, 4, 5, 6, 7, 8]，长度为 9
```

**思路**：（解法一 · 哈希集合，O(n)）利用集合的 O(1) 查找特性。遍历集合中的每个数 `num`，只有当 `num - 1` 不在集合中时，才从 `num` 开始向后查找连续序列（避免重复计算）。向右查找 `num + 1`、`num + 2`...直到不在集合中，更新最长长度。每个元素最多被访问两次，时间 O(n)；空间 O(n)。
（解法二 · 排序 + 指针，O(n log n)）先排序，再一次线性扫描：相邻相等跳过重复数字；`nums[right] == nums[right-1] + 1` 则连续，`cur += 1`；否则断开重置 `cur = 1`；全程 `ans` 维护最大值。排序主导时间 O(n log n)，空间 O(1)（原地排序）。

**亮点**：解法一用 `num - 1` 是否存在判断序列起点，每个连续序列只被遍历一次，将 O(n log n) 降至 O(n)。解法二最直观、是面试第一反应：先写出正确解，再追问「如何去掉排序」，自然引出解法一。

````python
from typing import List

class Solution:
    def longestConsecutive(self, nums: List[int]) -> int:
        # 利用集合快速查找某数是否在 nums 中
        num_set = set(nums)
        max_len = 0

        for num in num_set:
            # 只有连续序列的起点才会往后找（num - 1 不在集合中）
            if num - 1 not in num_set:
                cur = num
                cur_len = 1

                # 向后查找连续的数
                while cur + 1 in num_set:
                    cur += 1
                    cur_len += 1
                
                max_len = max(max_len, cur_len)
        
        return max_len

class Solution2:
    # 解法二：先排序再一次扫描。O(n log n) 不满足 O(n) 要求，但更直观、最容易第一时间想到
    def longestConsecutive(self, nums: List[int]) -> int:
        if not nums:
            return 0

        nums.sort()
        ans = 1
        cur = 1

        for right in range(1, len(nums)):
            # 重复数字跳过
            if nums[right] == nums[right - 1]:
                continue

            if nums[right] == nums[right - 1] + 1:
                # 连续
                cur += 1
            else:
                # 断开，重置
                cur = 1

            ans = max(ans, cur)

        return ans

# 测试
sol = Solution()
print(sol.longestConsecutive([100, 4, 200, 1, 3, 2]))  # 4
print(sol.longestConsecutive([0, 3, 7, 2, 5, 8, 4, 6, 0, 1]))  # 9
print(sol.longestConsecutive([]))  # 0
print(sol.longestConsecutive([1, 2, 0, 1]))  # 3

# 解法二测试
sol2 = Solution2()
print(sol2.longestConsecutive([100, 4, 200, 1, 3, 2]))  # 4
print(sol2.longestConsecutive([1, 2, 0, 1]))  # 3（重复的 1 被跳过）
print(sol2.longestConsecutive([]))  # 0
````

## 10. 等差数列划分

- **难度**：中等  
- **标签**：数组、双指针

**题目**：如果一个数列至少有三个元素，且任意两个相邻元素之差相同，则称该数列为等差数列。给定整数数组 `nums`，返回所有为等差数组的子数组个数。

**示例**：
```
输入：nums = [1, 2, 3, 4]
输出：3
解释：[1,2,3], [2,3,4], [1,2,3,4] 三个等差子数组

输入：nums = [1]
输出：0
```

**思路**：用增量法统计。遍历时检查 `nums[i] - nums[i-1] == nums[i-1] - nums[i-2]`。若成立，则以 `nums[i]` 结尾的等差子数组数比前一个位置多 1（`cur += 1`），累加到答案；若不成立则重置 `cur = 0`。时间 O(n)，空间 O(1)。

**亮点**：不需要统计连续段长度再算组合数，而是边走边累加增量，每个位置只关心「以自己结尾的新增子数组」，将问题转化为 O(n) 一次遍历。

````python
from typing import List

class Solution:
    def numberOfArithmeticSlices(self, nums: List[int]) -> int:
        ans = 0
        cur = 0
        for i in range(2, len(nums)):
            if nums[i] - nums[i-1] == nums[i-1] - nums[i-2]:
                cur += 1
                ans += cur
            else:
                cur = 0
        return ans

# 测试
sol = Solution()
print(sol.numberOfArithmeticSlices([1, 2, 3, 4]))  # 3
print(sol.numberOfArithmeticSlices([1]))  # 0
print(sol.numberOfArithmeticSlices([1, 2, 3, 4, 5]))  # 6
````

## 11. 接雨水

- **难度**：困难
- **标签**：双指针 / 动态规划 / 数组

**题目**：给定 n 个非负整数表示宽度为 1 的柱子的高度图，计算下雨后能接多少雨水。

**示例**：
```python
输入：height = [0,1,0,2,1,0,1,3,2,1,2,1]
输出：6

输入：height = [4,2,0,3,2,5]
输出：9
```

**思路**：双指针。每个位置 i 能接的水 = `min(leftMax, rightMax) - height[i]`。左右指针从两端向中间收敛，维护 `left_max`（[0..left] 最高）和 `right_max`（[right..n-1] 最高），每次移动较矮的一端：
- 若该端高度 ≥ 对应 max，更新 max；
- 否则该端可接水 = `max - height[i]`，累加到 ans。
时间 O(n)，空间 O(1)。

**亮点**：水量由较矮侧决定，故每次移动较矮端——此时该端水量只取决于同侧 max（对侧必然存在一根更高的柱子），可立即结算；把动态规划/单调栈的 O(n) 空间压到 O(1)。

````python
from typing import List

class Solution:
    def trap(self, height: List[int]) -> int:
        # water[i] = min(leftMax, rightMax) - height[i]
        # 双指针，移动低的

        left, right = 0, len(height) - 1
        left_max = right_max = 0
        ans = 0

        while left < right:
            if height[left] < height[right]:
                if height[left] >= left_max:
                    left_max = height[left]
                else:
                    ans += left_max - height[left]
                
                left += 1
            else:
                if height[right] >= right_max:
                    right_max = height[right]
                else:
                    ans += right_max - height[right]

                right -= 1
        return ans

# 测试
sol = Solution()
print(sol.trap([0,1,0,2,1,0,1,3,2,1,2,1]))  # 6
print(sol.trap([4,2,0,3,2,5]))               # 9
````
