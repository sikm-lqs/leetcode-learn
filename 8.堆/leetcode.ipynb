## 1. 前 K 个高频元素

- **难度**：中等
- **标签**：数组、哈希表、堆（优先队列）

**题目**：给定整数数组 `nums` 和整数 `k`，返回其中出现频率前 `k` 高的元素。可以按任意顺序返回答案。

**示例**：
```python
输入：nums = [1,1,1,2,2,3], k = 2
输出：[1, 2]
解释：元素 1 出现 3 次，元素 2 出现 2 次，元素 3 出现 1 次。前 2 高频元素是 [1, 2]

输入：nums = [1], k = 1
输出：[1]
```

**思路**：哈希表统计频率 + 小根堆维护前 k 个高频元素。

1. **统计频率**：用哈希表统计每个元素出现的次数
2. **堆维护**：用小根堆维护前 k 个高频元素
   - 将 [频率, 元素] 对入堆
   - 若堆大小超过 k，弹出堆顶（频率最低的）
3. **提取结果**：堆中剩余的就是前 k 个高频元素

**时间复杂度**：O(n log k) - n 次遍历统计 + n-k 次入堆/出堆  
**空间复杂度**：O(n) - 哈希表存储频率 + 堆空间 O(k)

**亮点**：
- 先用哈希表 O(n) 统计频率，避免每次排序
- 小根堆只保留前 k 个，时间复杂度 O(n log k) 优于全排序 O(n log n)
- 当 k << n 时，性能显著优于暴力排序法
- 代码简洁，利用 Python heapq 实现堆操作

````python
import heapq
from typing import List

class Solution:
    def topKFrequent(self, nums: List[int], k: int) -> List[int]:
        """
        哈希表统计频率 + 小根堆维护前 k 个高频元素
        时间复杂度：O(n log k)
        空间复杂度：O(n)
        """
        # 第1步：统计频率
        freq = {}
        for num in nums:
            freq[num] = freq.get(num, 0) + 1
        
        # 第2步：用小根堆维护前 k 个高频元素
        heap = []
        for num, count in freq.items():
            # 将 [频率, 元素] 对入堆
            heapq.heappush(heap, [count, num])
            
            # 如果堆大小超过 k，弹出频率最低的
            if len(heap) > k:
                heapq.heappop(heap)
        
        # 第3步：提取堆中的元素（只返回元素，不返回频率）
        return [item[1] for item in heap]

# 测试
sol = Solution()

# 测试用例 1
print(f"测试 1: {sol.topKFrequent([1,1,1,2,2,3], 2)}")    # 预期: [1, 2] 或 [2, 1]

# 测试用例 2
print(f"测试 2: {sol.topKFrequent([1], 1)}")              # 预期: [1]

# 测试用例 3
print(f"测试 3: {sol.topKFrequent([1,2,3,4,5], 2)}")       # 预期: [1, 2] 或其他两个

# 测试用例 4：所有元素频率相同
print(f"测试 4: {sol.topKFrequent([1,1,2,2,3,3], 2)}")    # 预期: 任意两个不同元素

# 测试用例 5：单个高频元素
print(f"测试 5: {sol.topKFrequent([5,5,5,5,1,2,3], 1)}")   # 预期: [5]
````

## 2. 数组中的第K个最大元素

- **难度**：中等
- **标签**：堆（优先队列）

**题目**：给定整数数组 `nums` 和整数 `k`，返回数组中第 `k` 个最大的元素（排序后的第 k 大，而非第 k 个不同元素）。

**示例**：
```python
输入：nums = [3,2,1,5,6,4], k = 2
输出：5
解释：升序为 [1,2,3,4,5,6]，第 2 大的元素是 5

输入：nums = [3,2,3,1,2,4,5,5,6], k = 4
输出：4
```

**思路**：用**大小为 k 的小根堆**维护当前最大的 k 个元素。遍历数组：
- 将元素加入堆
- 若堆大小超过 k，弹出堆顶（最小值）
- 最终堆顶就是第 k 大的元素

**核心思想**：小根堆始终保持当前最大的 k 个元素，堆顶是这 k 个中的最小值，即第 k 大。

**时间复杂度**：O(n log k) - n 次入堆，最多 k 次出堆  
**空间复杂度**：O(k) - 堆的大小

**亮点**：
- 相比全排序 O(n log n)，当 k << n 时更优
- 小根堆自动维护"最大的 k 个"，堆顶就是答案
- 代码简洁，易于理解和实现

````python
import heapq
from typing import List

class Solution:
    def findKthLargest(self, nums: List[int], k: int) -> int:
        """
        用大小为 k 的小根堆，维护当前最大的 k 个元素
        时间复杂度：O(n log k)
        空间复杂度：O(k)
        """
        # 创建小根堆
        heap = []
        
        for num in nums:
            # 将当前元素入堆
            heapq.heappush(heap, num)
            
            # 如果堆大小超过 k，弹出堆顶（最小的元素）
            # 这样堆中始终保持最大的 k 个元素
            if len(heap) > k:
                heapq.heappop(heap)
        
        # 堆顶就是第 k 大的元素
        return heap[0]

# 测试
sol = Solution()

# 测试用例 1
print(f"测试 1: {sol.findKthLargest([3,2,1,5,6,4], 2)}")      # 预期: 5

# 测试用例 2
print(f"测试 2: {sol.findKthLargest([3,2,3,1,2,4,5,5,6], 4)}") # 预期: 4

# 测试用例 3：k = 1（找最大值）
print(f"测试 3: {sol.findKthLargest([1,2,3,4,5], 1)}")        # 预期: 5

# 测试用例 4：k = n（找最小值）
print(f"测试 4: {sol.findKthLargest([1,2,3,4,5], 5)}")        # 预期: 1

# 测试用例 5：包含重复元素
print(f"测试 5: {sol.findKthLargest([2,2,2,2,2], 3)}")       # 预期: 2

# 测试用例 6：单个元素
print(f"测试 6: {sol.findKthLargest([1], 1)}")                  # 预期: 1

# 测试用例 7：负数
print(f"测试 7: {sol.findKthLargest([-1,-2,-3,-4,-5], 3)}")  # 预期: -3
````
