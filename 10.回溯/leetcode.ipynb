## 1. 括号生成

- **难度**：中等  
- **标签**：回溯、字符串

**题目**：数字 `n` 代表生成括号的对数，请你设计一个函数，用于能够生成所有可能的并且**有效**的括号组合。

**示例**：
```
输入：n = 3
输出：["((()))","(()())","(())()","()(())","()()()"]

输入：n = 1
输出：["()"]
```

**思路**：回溯（DFS）。`index` 记录已构造的字符数，`cnt` 记录当前未匹配的左括号数量。每个位置尝试两种选择：放 `(`（无条件），放 `)`（剪枝：仅当 `cnt > 0`，保证序列前缀永远合法）。构造满 `2n` 个字符且 `cnt == 0` 时记录答案。时间 O(C(2n, n))（卡特兰数级），空间 O(n)（递归深度 + 路径）。

**亮点**：用「未匹配左括号数 cnt」代替显式合法性校验——右括号只在有可配对左括号时才能添加，天然保证每个前缀都合法，无需最后验证；比「先生成再过滤」高效得多。


````python
from typing import List

class Solution:
    def generateParenthesis(self, n: int) -> List[str]:
        # 回溯
        res = []
        curr = [] # 表示现在正在构造的括号序列
        
        def dfs(index, cnt):
            # 构造了2n个字符
            if index == 2*n:
                # 未匹配的左括号数量为0，才合法
                if cnt == 0:
                    res.append("".join(curr))
                return

            curr.append("(")
            dfs(index+1, cnt+1)
            curr.pop()

            # 只有当前存在未匹配的左括号时，才可以添加右括号
            if cnt > 0:
                curr.append(")")
                dfs(index+1, cnt-1)
                curr.pop()
        
        dfs(0, 0) # 从第0个位置开始搜索，初始左括号数量为0
        return res

# 测试
sol = Solution()
print(sol.generateParenthesis(3))  # 预期: ["((()))","(()())","(())()","()(())","()()()"]
print(sol.generateParenthesis(1))  # 预期: ["()"]
````
