## 1. 任务调度器

- **难度**：中等
- **标签**：贪心、数组、哈希表

**题目**：给你一个用字符数组 `tasks` 表示的 CPU 需要执行的任务列表，用字母 A 到 Z 表示，以及一个冷却时间 `n`。每个周期或时间间隔允许完成一项任务。任务可以按任何顺序完成，但有一个限制：两个 **相同种类** 的任务之间必须有长度为 `n` 的冷却时间。

返回完成所有任务所需要的 **最短时间间隔**。

**示例**：
```python
输入：tasks = ["A","A","A","B","B","B"], n = 2
输出：8
解释：一种可能的执行顺序是：A -> B -> idle -> A -> B -> idle -> A -> B

输入：tasks = ["A","A","A","B","B","B"], n = 0
输出：6
解释：可以直接按 A -> A -> A -> B -> B -> B 执行
```

**思路**：贪心策略 —— 优先处理数量最多、最难安排的任务，并尽可能用其他任务填满它造成的冷却空隙。

关键公式：`最短时间 = max(len(tasks), (maxFreq - 1) * (n + 1) + maxCount)`

- **maxFreq**：出现频率最高的任务的次数
- **maxCount**：频率最高的任务有多少个不同的种类
- **(maxFreq - 1) * (n + 1)**：最后一个高频任务前的框架长度
- **+ maxCount**：加上最后一轮的高频任务数量

**时间复杂度**：O(n) - 统计频率 + 计算最大值
**空间复杂度**：O(1) - 哈希表最多26个字母

**亮点**：巧妙地将调度问题转化为数学计算，通过贪心思想找到最优调度策略。

````python
from typing import List

class Solution:
    def leastInterval(self, tasks: List[str], n: int) -> int:
        """
        贪心：优先处理数量最多、最难安排的任务，并尽可能用其他任务填满它造成的冷却空隙。
        时间复杂度：O(n)
        空间复杂度：O(1)
        """
        # 统计每个任务的频率
        cnt = {}
        for task in tasks:
            if task not in cnt:
                cnt[task] = 1
            else:
                cnt[task] += 1
        
        # 找到最高频率和最高频率的任务数量
        max_freq = max(cnt.values())
        max_count = sum(freq == max_freq for freq in cnt.values())
        
        # 计算最短时间间隔
        # (max_freq - 1) * (n + 1) + max_count：框架长度
        # len(tasks)：如果没有空闲时间，直接返回任务数
        return max(len(tasks), (max_freq - 1) * (n + 1) + max_count)

# 测试
sol = Solution()

# 测试用例 1：标准情况
print(f"测试 1: {sol.leastInterval(['A','A','A','B','B','B'], 2)}")   # 预期: 8

# 测试用例 2：无冷却时间
print(f"测试 2: {sol.leastInterval(['A','A','A','B','B','B'], 0)}")   # 预期: 6

# 测试用例 3：冷却时间很大
print(f"测试 3: {sol.leastInterval(['A','A','A','B','B','B'], 3)}")   # 预期: 10

# 测试用例 4：多种任务
print(f"测试 4: {sol.leastInterval(['A','A','A','B','B','C','C','D','D'], 2)}")  # 预期: 9

# 测试用例 5：单个任务
print(f"测试 5: {sol.leastInterval(['A','A','A'], 2)}")             # 预期: 7

# 测试用例 6：所有任务不同
print(f"测试 6: {sol.leastInterval(['A','B','C','D','E'], 3)}")      # 预期: 5
````

## 2. 跳跃游戏

- **难度**：中等
- **标签**：贪心、数组

**题目**：给你一个非负整数数组 `nums`，你最初位于数组的 **第一个下标**。数组中的每个元素代表你在该位置可以 **跳跃的最大长度**。

判断你是否能够到达最后一个下标，如果可以，返回 `true`；否则，返回 `false`。

**示例**：
```python
输入：nums = [2,3,1,1,4]
输出：true
解释：可以先跳 1 步，从下标 0 到达下标 1，然后从下标 1 跳 3 步到达最后一个下标。

输入：nums = [3,2,1,0,4]
输出：false
解释：无论怎样，总会到达下标 3 的位置，但该位置的最大跳跃长度是 0，无法到达最后一个下标。
```

**思路**：贪心策略 —— 维护 `max_reach`（当前能够到达的最远距离），每一步贪心地扩大这个覆盖范围。

- 遍历数组，维护 `max_reach = max(max_reach, i + nums[i])`
- 如果当前索引 `i > max_reach`，说明无法到达该位置，返回 `false`
- 如果能够遍历完整个数组，说明可以到达最后一个位置，返回 `true`

**时间复杂度**：O(n) - 只需遍历一次
**空间复杂度**：O(1) - 只用常数空间

**亮点**：通过维护一个不断扩大的覆盖范围，将复杂的问题简化为简单的边界判断。

````python
from typing import List

class Solution:
    def canJump(self, nums: List[int]) -> bool:
        """
        贪心：维护 max_reach，当前所能到达的最远距离。
        每一步维护目前能够达到的最大覆盖范围，并贪心地扩大这个范围。
        时间复杂度：O(n)
        空间复杂度：O(1)
        """
        max_reach = 0
        
        for i in range(len(nums)):
            # 如果当前索引无法到达，返回 false
            if i > max_reach:
                return False
            
            # 更新能够到达的最远距离
            max_reach = max(max_reach, i + nums[i])
        
        return True

# 测试
sol = Solution()

# 测试用例 1：可以到达
print(f"测试 1: {sol.canJump([2,3,1,1,4])}")                 # 预期: True

# 测试用例 2：无法到达
print(f"测试 2: {sol.canJump([3,2,1,0,4])}")                 # 预期: False

# 测试用例 3：单个元素
print(f"测试 3: {sol.canJump([0])}")                        # 预期: True

# 测试用例 4：从开头就能跳到终点
print(f"测试 4: {sol.canJump([5,0,0,0,0])}")                 # 预期: True

# 测试用例 5：需要多步跳跃
print(f"测试 5: {sol.canJump([1,1,1,1,1])}")                 # 预期: True

# 测试用例 6：零阻碍
print(f"测试 6: {sol.canJump([1,0,1,0,1])}")                 # 预期: False
````

## 3. 买卖股票的最佳时机

- **难度**：简单
- **标签**：贪心、数组

**题目**：给定一个数组 `prices`，它的第 `i` 个元素 `prices[i]` 表示一支给定股票第 `i` 天的价格。

你只能选择 **某一天** 买入这只股票，并选择在 **未来的某一个不同的日子** 卖出该股票。设计一个算法来计算你所能获取的最大利润。

返回你可以从这笔交易中获取的最大利润。如果你不能获取任何利润，返回 `0`。

**示例**：
```python
输入：prices = [7,1,5,3,6,4]
输出：5
解释：在第 2 天（股票价格 = 1）的时候买入，在第 5 天（股票价格 = 6）的时候卖出，最大利润 = 6 - 1 = 5。

输入：prices = [7,6,4,3,1]
输出：0
解释：在这种情况下，没有交易完成，所以最大利润为 0。
```

**思路**：贪心策略 —— 遍历每一天卖出，利润最大的买入时机一定是在之前价格最低的那天。

- 维护 `min_price`：遍历过程中的最低价格
- 维护 `max_profit`：遍历过程中的最大利润
- 对于每一天，计算 `当前利润 = 当天价格 - 历史最低价格`
- 更新最大利润和最低价格

**时间复杂度**：O(n) - 只需遍历一次
**空间复杂度**：O(1) - 只用常数空间

**亮点**：将问题转化为"在遍历过程中维护历史最低值"，一次遍历即可得到最优解。

````python
from typing import List

class Solution:
    def maxProfit(self, prices: List[int]) -> int:
        """
        贪心：遍历每天卖出，利润最大的买的时间一定是在（1, i-1）中 prices[x] 最低的那天。
        时间复杂度：O(n)
        空间复杂度：O(1)
        """
        if not prices or len(prices) < 2:
            return 0
        
        min_price = prices[0]
        max_profit = 0
        
        for i in range(1, len(prices)):
            # 计算如果在第 i 天卖出的利润
            cur_profit = prices[i] - min_price
            
            # 更新最大利润
            max_profit = max(max_profit, cur_profit)
            
            # 更新最低价格
            min_price = min(min_price, prices[i])
        
        return max_profit if max_profit > 0 else 0

# 测试
sol = Solution()

# 测试用例 1：标准情况
print(f"测试 1: {sol.maxProfit([7,1,5,3,6,4])}")             # 预期: 5

# 测试用例 2：价格一直下跌
print(f"测试 2: {sol.maxProfit([7,6,4,3,1])}")              # 预期: 0

# 测试用例 3：价格一直上涨
print(f"测试 3: {sol.maxProfit([1,2,3,4,5])}")              # 预期: 4

# 测试用例 4：两天
print(f"测试 4: {sol.maxProfit([2,4])}")                     # 预期: 2

# 测试用例 5：先跌后涨
print(f"测试 5: {sol.maxProfit([3,2,6,5,0,3])}")             # 预期: 4

# 测试用例 6：单一价格
print(f"测试 6: {sol.maxProfit([5,5,5,5])}")                # 预期: 0
````
