## 1. 搜索插入位置

- **难度**：简单
- **标签**：数组、二分查找

**题目**：给定一个排序数组和一个目标值，在数组中找到目标值，并返回其索引。如果目标值不存在于数组中，返回它将会被按顺序插入的位置。

**示例**：
```python
输入：nums = [1,3,5,6], target = 5
输出：2

输入：nums = [1,3,5,6], target = 2
输出：1
```

**思路**：使用左闭右闭区间 [left, right] 进行二分查找。初始化 left = 0, right = len(nums)-1。
当 left <= right 时，mid = (left + right) // 2。
若 nums[mid] < target 则 left = mid + 1，否则 right = mid - 1。
循环结束时 left 即为第一个 >= target 的位置（插入位置）。

**时间复杂度**：O(log n)
**空间复杂度**：O(1)

**亮点**：左闭右闭的经典写法，通过 right = mid-1 收缩范围，最终 left 自然指向插入位置，是最直观的二分查找模板。

````python
from typing import List

class Solution:
    def searchInsert(self, nums: List[int], target: int) -> int:
        left, right = 0, len(nums) - 1
        
        while left <= right:
            mid = (left + right) // 2
            
            if nums[mid] < target:
                left = mid + 1
            else:
                right = mid - 1
        
        return left

# 测试
sol = Solution()

# 测试用例 1：目标存在
print(f"测试 1: {sol.searchInsert([1, 3, 5, 6], 5)}")      # 预期: 2

# 测试用例 2：目标不存在（应插入到位置1）
print(f"测试 2: {sol.searchInsert([1, 3, 5, 6], 2)}")      # 预期: 1 (插入位置)

# 测试用例 3：目标大于所有元素（应插入到末尾）
print(f"测试 3: {sol.searchInsert([1, 3, 5, 6], 7)}")      # 预期: 4

# 测试用例 4：目标小于所有元素（应插入到开头）
print(f"测试 4: {sol.searchInsert([1, 3, 5, 6], 0)}")      # 预期: 0

# 测试用例 5：空数组
print(f"测试 5: {sol.searchInsert([], 5)}")                    # 预期: 0
````

## 2. 在排序数组中查找元素的第一个和最后一个位置

- **难度**：中等
- **标签**：数组、二分查找

**题目**：给你一个按照非递减顺序排列的整数数组 nums，和一个目标值 target。找出给定目标值在数组中的开始位置和结束位置。如果数组中不存在目标值 target，返回 [-1, -1]。

**示例**：
```python
输入：nums = [5,7,7,8,8,10], target = 8
输出：[3, 4]

输入：nums = [5,7,7,8,8,10], target = 6
输出：[-1, -1]

输入：nums = [], target = 0
输出：[-1, -1]
```

**思路**：两次二分，分别找到 target 的左边界和右边界。

- **找左边界**：标准的二分查找，当 nums[mid] == target 时，继续向左查找（right = mid - 1）
- **找右边界**：类似的二分查找，当 nums[mid] == target 时，继续向右查找（left = mid + 1）

**时间复杂度**：O(log n) - 两次二分查找
**空间复杂度**：O(1)

**亮点**：将一个问题分解为两个独立的二分查找，分别找左边界和右边界。通过修改二分方向（mid + 1 或 mid - 1）来决定继续向左或向右搜索。

````python
from typing import List

class Solution:
    def searchRange(self, nums: List[int], target: int) -> List[int]:
        """
        两次二分查找，分别找到target的左边界和右边界
        时间复杂度：O(log n)
        空间复杂度：O(1)
        """
        
        def find_left():
            left, right = 0, len(nums) - 1
            ans = -1
            
            while left <= right:
                mid = (left + right) // 2
                
                if nums[mid] == target:
                    ans = mid
                    right = mid - 1  # 继续往左找
                elif nums[mid] < target:
                    left = mid + 1
                else:
                    right = mid - 1
            
            return ans
        
        def find_right():
            left, right = 0, len(nums) - 1
            ans = -1
            
            while left <= right:
                mid = (left + right) // 2
                
                if nums[mid] == target:
                    ans = mid
                    left = mid + 1  # 继续往右找
                elif nums[mid] < target:
                    left = mid + 1
                else:
                    right = mid - 1
            
            return ans
        
        return [find_left(), find_right()]

# 测试
sol = Solution()

# 测试用例 1：目标存在
print(f"测试 1: {sol.searchRange([5,7,7,8,8,10], 8)}")       # 预期: [3, 4]

# 测试用例 2：目标不存在
print(f"测试 2: {sol.searchRange([5,7,7,8,8,10], 6)}")       # 预期: [-1, -1]

# 测试用例 3：单个目标元素
print(f"测试 3: {sol.searchRange([5,7,7,8,8,10], 5)}")       # 预期: [0, 0]

# 测试用例 4：目标在中间位置
print(f"测试 4: {sol.searchRange([1,2,3,3,3,3,4], 3)}")       # 预期: [2, 4]

# 测试用例 5：空数组
print(f"测试 5: {sol.searchRange([], 0)}")                      # 预期: [-1, -1]
````

## 3. 搜索二维矩阵

- **难度**：中等
- **标签**：数组、二分查找、矩阵

**题目**：给你一个满足下述两条属性的 m x n 整数矩阵：
- 每行中的整数从左到右按非严格递增顺序排列
- 每行的第一个整数大于前一行的最后一个整数

给定一个整数 target，如果 target 在矩阵中，返回 true；否则，返回 false。

**示例**：
```python
输入：matrix = [
  [1,   3,  5,  7],
  [10, 11, 16, 20],
  [23, 30, 34, 60]
], target = 3
输出：true

输入：matrix = 同上, target = 13
输出：false
```

**思路**：将二维矩阵看作一个排好序的一维数组，对索引进行二分查找。核心是将一维数组的 mid 索引转换为二维坐标：
- i = mid // n（行索引）
- j = mid % n（列索引）

**时间复杂度**：O(log(m * n)) - 二分范围是 m*n
**空间复杂度**：O(1)

**亮点**：巧妙地将二维矩阵转化为一维数组问题，利用行列单调性，只需一次二分查找即可定位目标值。

````python
from typing import List

class Solution:
    def searchMatrix(self, matrix: List[List[int]], target: int) -> bool:
        """
        将二维矩阵看作一维数组，进行二分查找
        时间复杂度：O(log(m * n))
        空间复杂度：O(1)
        """
        if not matrix or not matrix[0]:
            return False
        
        m, n = len(matrix), len(matrix[0])
        left, right = 0, m * n - 1
        
        while left <= right:
            mid = (left + right) // 2
            
            # 一维坐标转化为二维坐标
            i = mid // n
            j = mid % n
            value = matrix[i][j]
            
            if value == target:
                return True
            elif value < target:
                left = mid + 1
            else:
                right = mid - 1
        
        return False

# 测试
sol = Solution()

# 测试用例 1：目标存在
matrix1 = [
  [1, 3, 5, 7],
  [10, 11, 16, 20],
  [23, 30, 34, 60]
]
print(f"测试 1: {sol.searchMatrix(matrix1, 3)}")             # 预期: True

# 测试用例 2：目标不存在
print(f"测试 2: {sol.searchMatrix(matrix1, 13)}")            # 预期: False

# 测试用例 3：目标在边界
print(f"测试 3: {sol.searchMatrix(matrix1, 1)}")             # 预期: True
print(f"测试 4: {sol.searchMatrix(matrix1, 60)}")            # 预期: True

# 测试用例 4：单行矩阵
matrix2 = [[1, 3, 5, 7, 9]]
print(f"测试 5: {sol.searchMatrix(matrix2, 5)}")             # 预期: True

# 测试用例 5：单元素矩阵
matrix3 = [[5]]
print(f"测试 6: {sol.searchMatrix(matrix3, 5)}")             # 预期: True
print(f"测试 7: {sol.searchMatrix(matrix3, 1)}")             # 预期: False

# 测试用例 6：空矩阵
matrix4 = []
print(f"测试 8: {sol.searchMatrix(matrix4, 5)}")              # 预期: False
````

## 4. 搜索旋转排序数组

- **难度**：中等
- **标签**：数组、二分查找

**题目**：整数数组 nums 按升序排列，数组中的值 互不相同。在传递给函数之前，nums 在预先未知的某个下标 k（0 <= k < nums.length）上进行了向左旋转，使数组变为 [nums[k], nums[k+1], ..., nums[n-1], nums[0], nums[1], ..., nums[k-1]]。给定旋转后的数组 nums 和一个整数 target，如果 nums 中存在这个目标值 target，则返回它的下标，否则返回 -1。

**示例**：
```python
输入：nums = [4,5,6,7,0,1,2], target = 0
输出：4

输入：nums = [4,5,6,7,0,1,2], target = 3
输出：-1

输入：nums = [1], target = 0
输出：0
```

**思路**：旋转数组有一个关键特性：至少有一半是有序的。二分时先判断哪一半是有序的，然后判断 target 是否在有序的那一半内。

- 若 nums[left] <= nums[mid]：左半边有序
  - 如果 nums[left] <= target <= nums[mid]，target 在左半边，right = mid - 1
  - 否则 target 在右半边，left = mid + 1
- 若 nums[left] > nums[mid]：右半边有序
  - 如果 nums[mid] <= target <= nums[right]，target 在右半边，left = mid + 1
  - 否则 target 在左半边，right = mid - 1

**时间复杂度**：O(log n) - 每次将搜索范围减半
**空间复杂度**：O(1)

**亮点**：利用旋转数组「一半有序」的特性，每次二分都能排除一半的搜索空间。关键是通过边界判断哪一半是有序的。

````python
from typing import List

class Solution:
    def search(self, nums: List[int], target: int) -> int:
        """
        在旋转排序数组中搜索目标值
        时间复杂度：O(log n)
        空间复杂度：O(1)
        """
        left, right = 0, len(nums) - 1
        
        while left <= right:
            mid = (left + right) // 2
            
            # 找到目标值
            if nums[mid] == target:
                return mid
            
            # 判断哪一半是有序的
            if nums[left] <= nums[mid]:
                # 左半边有序
                if nums[left] <= target <= nums[mid]:
                    right = mid - 1
                else:
                    left = mid + 1
            else:
                # 右半边有序
                if nums[mid] <= target <= nums[right]:
                    left = mid + 1
                else:
                    right = mid - 1
        
        return -1

# 测试
sol = Solution()

# 测试用例 1：目标存在
print(f"测试 1: {sol.search([4,5,6,7,0,1,2], 0)}")        # 预期: 4

# 测试用例 2：目标不存在
print(f"测试 2: {sol.search([4,5,6,7,0,1,2], 3)}")        # 预期: -1

# 测试用例 3：目标在左半边
print(f"测试 3: {sol.search([6,7,0,1,2,4,5], 6)}")        # 预期: 0

# 测试用例 4：目标在右半边
print(f"测试 4: {sol.search([4,5,6,7,0,1,2], 3)}")        # 预期: 1

# 测试用例 5：单个元素
print(f"测试 5: {sol.search([1], 1)}")                       # 预期: 0
print(f"测试 6: {sol.search([1], 0)}")                       # 预期: -1

# 测试用例 6：两个元素
print(f"测试 7: {sol.search([1,3], 3)}")                       # 预期: 1
````
