## 1. 有效的括号

- **难度**：简单  
- **标签**：栈

**题目**：给定一个只包括 `(`、`)`、`{`、`}`、`[`、`]` 的字符串 `s`，判断字符串是否有效。有效字符串需满足左括号必须用相同类型的右括号闭合，且左括号必须以正确的顺序闭合。

**示例**：```
输入：s = "()"
输出：true

输入：s = "()[]{}"
输出：true

输入：s = "(]"
输出：false
```

**思路**：用栈存储左括号，遇到右括号时检查栈顶是否匹配。遍历过程中若出现不匹配或栈为空则提前返回 false，最后检查栈是否为空。时间 O(n)，空间 O(n)。

**亮点**：栈的典型应用——利用后进先出特性匹配括号，匹配关系用哈希表存储使代码更简洁。

````python
from typing import List

class Solution:
    def isValid(self, s: str) -> bool:
        stack = []
        pairs = {"(": ")", "[": "]", "{": "}"}
        for char in s:
            if char in "([{":
                stack.append(char)
            else:
                if not stack or char != pairs[stack.pop()]:
                    return False
        return not stack

# 测试
sol = Solution()
print(sol.isValid("()"))  # True
print(sol.isValid("()[]{}"))  # True
print(sol.isValid("(]"))  # False
print(sol.isValid("([)]"))  # False
print(sol.isValid("{[]}"))  # True
````

## 2. 最小栈

- **难度**：中等  
- **标签**：栈、设计

**题目**：设计一个支持 `push`、`pop`、`top` 操作，并能在常数时间内检索到最小元素的栈。

**示例**：```
输入：
["MinStack","push","push","push","getMin","pop","top","getMin"]
[[],[-2],[0],[-3],[],[],[],[]]
输出：[null,null,null,null,-3,null,0,-2]
```

**思路**：维护两个栈：数据栈正常存元素，辅助栈同步存当前最小值。push 时辅助栈压入 `min(value, 辅助栈栈顶)`；pop 时两个栈同步弹出。时间 O(1)，空间 O(n)。

**亮点**：用辅助栈以空间换时间，每个状态下的最小值都被保留，不会因为 pop 而丢失历史最小值。

````python
from typing import List

class MinStack:
    def __init__(self):
        self.stack = []
        self.min_stack = []

    def push(self, value: int) -> None:
        self.stack.append(value)
        if not self.min_stack:
            self.min_stack.append(value)
        else:
            self.min_stack.append(min(value, self.min_stack[-1]))

    def pop(self) -> None:
        self.stack.pop()
        self.min_stack.pop()

    def top(self) -> int:
        return self.stack[-1]

    def getMin(self) -> int:
        return self.min_stack[-1]

# 测试
obj = MinStack()
obj.push(-2)
obj.push(0)
obj.push(-3)
print(obj.getMin())  # -3
obj.pop()
print(obj.top())     # 0
print(obj.getMin())  # -2
````

## 3. 字符串解码

- **难度**：中等  
- **标签**：栈、字符串

**题目**：给定一个经过编码的字符串 `s`，返回解码后的字符串。编码规则为 `k[encoded_string]`，表示方括号内部的字符串正好重复 `k` 次。

**示例**：```
输入：s = "3[a]2[bc]"
输出："aaabcbc"

输入：s = "3[a2[c]]"
输出："accaccacc"

输入：s = "2[abc]3[cd]ef"
输出："abcabccdcdcdef"
```

**思路**：用栈处理嵌套。遍历字符：非 `]` 入栈；遇到 `]` 则弹出字符直到 `[`，再弹出数字确定重复次数，将解码后的字符串重新入栈。最后将栈中所有字符串拼接。时间 O(n)，空间 O(n)。

**亮点**：栈嵌套解码——遇到 `]` 时弹出括号内内容、数字、再压回，利用栈天然处理嵌套结构，无需递归。

````python
from typing import List

class Solution:
    def decodeString(self, s: str) -> str:
        stack = []
        for char in s:
            if char != "]":
                stack.append(char)
            else:
                cur = ""
                while stack[-1] != "[":
                    cur = stack.pop() + cur
                stack.pop()  # 弹出 "["
                num = ""
                while stack and stack[-1].isdigit():
                    num = stack.pop() + num
                stack.append(int(num) * cur)
        return "".join(stack)

# 测试
sol = Solution()
print(sol.decodeString("3[a]2[bc]"))  # "aaabcbc"
print(sol.decodeString("3[a2[c]]"))   # "accaccacc"
print(sol.decodeString("2[abc]3[cd]ef"))  # "abcabccdcdcdef"
print(sol.decodeString("abc3[cd]xyz"))    # "abccdcdcdxyz" 
````

## 4. 每日温度

- **难度**：中等  
- **标签**：栈、单调栈

**题目**：给定整数数组 `temperatures`，返回数组 `answer`，其中 `answer[i]` 指第 `i` 天后下一个更高温度出现在几天后。若之后不会升高，则用 0 代替。

**示例**：```
输入：temperatures = [73,74,75,71,69,72,76,73]
输出：[1,1,4,2,1,1,0,0]

输入：temperatures = [30,40,50,60]
输出：[1,1,1,0]

输入：temperatures = [30,60,90]
输出：[1,1,0]
```

**思路**：单调递减栈。遍历温度，栈中存未找到更高温度的下标（从栈底到栈顶温度递减）。当前温度高于栈顶下标对应温度时，不断弹出并计算天数差。时间 O(n)，空间 O(n)。

**亮点**：单调栈的经典应用——每个元素最多入栈出栈一次，将暴力 O(n²) 降至 O(n)；栈中保持递减确保找到的「下一个更高」就是最近的。

````python
from typing import List

class Solution:
    def dailyTemperatures(self, temperatures: List[int]) -> List[int]:
        n = len(temperatures)
        answer = [0] * n
        stack = []
        for i, temp in enumerate(temperatures):
            while stack and temp > temperatures[stack[-1]]:
                prev = stack.pop()
                answer[prev] = i - prev
            stack.append(i)
        return answer

# 测试
sol = Solution()
print(sol.dailyTemperatures([73,74,75,71,69,72,76,73]))  # [1,1,4,2,1,1,0,0]
print(sol.dailyTemperatures([30,40,50,60]))              # [1,1,1,0]
print(sol.dailyTemperatures([30,60,90]))                 # [1,1,0]
````
