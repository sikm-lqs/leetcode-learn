## 1. 有效的括号

- **难度**：简单  
- **标签**：栈

**题目**：给定一个只包括 '('，')'，'{'，'}'，'['，']' 的字符串 s ，判断字符串是否有效。

有效字符串需满足：

- 左括号必须用相同类型的右括号闭合。
- 左括号必须以正确的顺序闭合。
- 每个右括号都有一个对应的相同类型的左括号。

**示例**：
```
输入：s = "()"
输出：true

输入：s = "()[]{}"
输出：true

输入：s = "(]"
输出：false
```

**思路**：利用栈的先进后出。遍历字符串：
- 遇到左括号就入栈
- 遇到右括号就检查栈顶是否匹配对应左括号（用 map 存储映射）
若不匹配或栈为空则返回 false。遍历完后栈必须为空。

时间复杂度：O(n)
空间复杂度：O(n)

**亮点**：栈的 LIFO 特性完美解决括号匹配问题，用哈希表存映射使代码清晰简洁，是栈的经典应用题。

````python
from typing import List

class Solution:
    def isValid(self, s: str) -> bool:
        # 利用栈的先进后出
        stack = []

        map = {')':'(',']':'[','}':'{'}

        for char in s:
            # 左括号就入栈
            if char in '([{' :
                stack.append(char)
            
            # 右括号就检查是否匹配        
            elif char in ')]}':
                if not stack or map[char] != stack.pop():
                    return False
            
        return len(stack) == 0
        

# 测试
sol = Solution()
print(sol.isValid("()"))  # True
print(sol.isValid("()[]{}"))  # True
print(sol.isValid("(]"))  # False
print(sol.isValid("([)]"))  # False
print(sol.isValid("{[]}"))  # True
````

## 2. 最小栈

- **难度**：中等  
- **标签**：栈、设计

**题目**：设计一个支持 `push`、`pop`、`top` 操作，并能在常数时间内检索到最小元素的栈。

**示例**：```
输入：
["MinStack","push","push","push","getMin","pop","top","getMin"]
[[],[-2],[0],[-3],[],[],[],[]]
输出：[null,null,null,null,-3,null,0,-2]
```

**思路**：维护两个栈：数据栈正常存元素，辅助栈同步存当前最小值。push 时辅助栈压入 `min(value, 辅助栈栈顶)`；pop 时两个栈同步弹出。时间 O(1)，空间 O(n)。

**亮点**：用辅助栈以空间换时间，每个状态下的最小值都被保留，不会因为 pop 而丢失历史最小值。

````python
from typing import List

class MinStack:

    def __init__(self):
        self.data_stack = []
        self.min_stack = []

    
    def push(self, value: int) -> None:
        self.data_stack.append(value)
        # 压入最小栈：最小栈为空或者当前元素小于等于最小栈栈顶
        if not self.min_stack or value <= self.min_stack[-1]:
            self.min_stack.append(value)
        

    def pop(self) -> None:
        top_element = self.data_stack.pop()

        # 如果弹出元素等于最小栈栈顶，最小栈也要同步弹出
        if top_element == self.min_stack[-1]:
            self.min_stack.pop()
        

    def top(self) -> int:
        return self.data_stack[-1]
        

    def getMin(self) -> int:
        return self.min_stack[-1]

# 测试
obj = MinStack()
obj.push(-2)
obj.push(0)
obj.push(-3)
print(obj.getMin())  # -3
obj.pop()
print(obj.top())     # 0
print(obj.getMin())  # -2
````

## 3. 字符串解码

- **难度**：中等  
- **标签**：栈、字符串

**题目**：给定一个经过编码的字符串 `s`，返回解码后的字符串。编码规则为 `k[encoded_string]`，表示方括号内部的字符串正好重复 `k` 次。

**示例**：```
输入：s = "3[a]2[bc]"
输出："aaabcbc"

输入：s = "3[a2[c]]"
输出："accaccacc"

输入：s = "2[abc]3[cd]ef"
输出："abcabccdcdcdef"
```

**思路**：用栈处理嵌套。遍历字符：非 `]` 入栈；遇到 `]` 则弹出字符直到 `[`，再弹出数字确定重复次数，将解码后的字符串重新入栈。最后将栈中所有字符串拼接。时间 O(n)，空间 O(n)。

**亮点**：栈嵌套解码——遇到 `]` 时弹出括号内内容、数字、再压回，利用栈天然处理嵌套结构，无需递归。

````python
from typing import List

class Solution:
    def decodeString(self, s: str) -> str:
        # 使用栈模拟解码过程
        stack = []

        # 遍历字符串中的每个字符
        for c in s:
            # 遇到非右括号字符，直接入栈
            if c != "]":
                stack.append(c)
            else:
                # 1. 取出括号中的字符串
                now_str = ""
                while stack[-1] != "[":
                    now_str += stack[-1]
                    stack.pop()

                # 弹出左括号
                stack.pop()

                # 因为是从栈顶取出的，所以需要反转
                now_str = now_str[::-1]

                # 2. 取出左括号前面的数字
                num_str = ""
                while stack and stack[-1].isdigit():
                    num_str += stack[-1]
                    stack.pop()

                # 数字也是倒着取出的，所以需要反转
                num_str = num_str[::-1]
                num = int(num_str)

                # 3. 将解码后的字符串重新压入栈中
                for _ in range(num):
                    for ch in now_str:
                        stack.append(ch)

        # 栈中剩余字符拼接成最终结果
        return "".join(stack)


# 测试
sol = Solution()
print(sol.decodeString("3[a]2[bc]"))  # "aaabcbc"
print(sol.decodeString("3[a2[c]]"))   # "accaccacc"
print(sol.decodeString("2[abc]3[cd]ef"))  # "abcabccdcdcdef"
print(sol.decodeString("abc3[cd]xyz"))    # "abccdcdcdxyz" 
````

## 4. 每日温度

- **难度**：中等  
- **标签**：栈、单调栈

**题目**：给定整数数组 `temperatures`，返回数组 `answer`，其中 `answer[i]` 指第 `i` 天后下一个更高温度出现在几天后。若之后不会升高，则用 0 代替。

**示例**：```
输入：temperatures = [73,74,75,71,69,72,76,73]
输出：[1,1,4,2,1,1,0,0]

输入：temperatures = [30,40,50,60]
输出：[1,1,1,0]

输入：temperatures = [30,60,90]
输出：[1,1,0]
```

**思路**：使用单调栈维护一批还没有被解决的问题（即还没有找到下一个更高温度的日子）。

遍历每个温度：
- 当当前温度比栈顶温度高时，说明找到了栈顶那些未解决天的答案，批量弹出并记录天数差（i - prev）
- 否则将当前下标入栈

最后栈中剩余的元素没有更高温度，答案保持为 0。

时间复杂度：O(n)
空间复杂度：O(n)

**亮点**：单调栈的精髓是批量解决未完成的问题。每个元素只进出栈一次，从 O(n^2) 暴力优化到 O(n)。


````python
from typing import List

class Solution:
    def dailyTemperatures(self, temperatures: List[int]) -> List[int]:
        # 单调栈
        # 维护一批还没有被解决的问题，当未来出现答案时批量解决。
        n = len(temperatures)
        ans = [0] * n

        stack = []  # 存下标

        for i, t in enumerate(temperatures):
            # 当前温度比栈顶温度高
            while stack and t > temperatures[stack[-1]]:
                prev = stack.pop()
                ans[prev] = i - prev

            stack.append(i)

        return ans

# 测试
sol = Solution()
print(sol.dailyTemperatures([73,74,75,71,69,72,76,73]))  # [1,1,4,2,1,1,0,0]
print(sol.dailyTemperatures([30,40,50,60]))              # [1,1,1,0]
print(sol.dailyTemperatures([30,60,90]))                 # [1,1,0]
````

## 5. 简化路径

- **难度**：中等
- **标签**：栈 / 字符串

**题目**：给你一个字符串 path，表示指向某一文件或目录的 Unix 风格绝对路径（以 '/' 开头），请你将其转化为更加简洁的规范路径。

**Unix 风格文件系统规则**：
- 一个点 `.` 表示当前目录本身
- 两个点 `..` 表示将目录切换到上一级（指向父目录）
- 任意多个连续的斜杠（即 `'//'` 或 `'///``）都被视为单个斜杠 `'/'`
- 任何其他格式的点（例如 `'...'` 或 `'....'`）均被视为有效的文件/目录名称

**规范路径格式**：
- 始终以斜杠 `'/'` 开头
- 两个目录名之间必须只有一个斜杠 `'/'`
- 最后一个目录名（如果存在）不能以 `'/'` 结尾
- 路径仅包含从根目录到目标文件或目录的路径（不含 `'.'` 或 `'..'`）

**示例**：
```python
输入：path = "/home/"
输出："/home"

输入：path = "/../"
输出："/"

输入：path = "/home//foo/"
输出："/home/foo"

输入：path = "/a/./b/../../c/"
输出："/c"
```

**思路**：用栈模拟目录切换。按 `'/'` 分割路径，逐个处理：
- 空串或 `'.'` → 忽略（当前目录）
- `'..'` → 弹出栈顶（返回上一级，若栈非空）
- 其他 → 入栈（目录名）

最后用 `'/'` 连接栈中元素。时间 O(n)，空间 O(n)。

**亮点**：栈模拟目录层次结构，`'..'` 对应出栈，目录名对应入栈；利用 `split('/')` 自动处理连续斜杠，简化逻辑。

````python
from typing import List

class Solution:
    def simplifyPath(self, path: str) -> str:
        """
        用栈模拟目录切换
        时间复杂度：O(n)
        空间复杂度：O(n)
        """
        stack = []

        # 按照'/'分割路径
        for name in path.split("/"):
            # 空串与.直接忽略
            if name == "" or name == ".":
                continue
            # ..返回上一级
            elif name == "..":
                if stack:
                    stack.pop()
            # 普通目录名入栈
            else:
                stack.append(name)

        # 栈为空说明是根目录
        if not stack:
            return '/'        
        
        return '/' + '/'.join(stack)

# 测试
sol = Solution()

# 测试用例 1：普通路径
print(f"测试 1: {sol.simplifyPath('/home/')}")              # 预期: "/home"
print(f"测试 2: {sol.simplifyPath('/../')}")                # 预期: "/"

# 测试用例 2：连续斜杠
print(f"测试 3: {sol.simplifyPath('/home//foo/')}")         # 预期: "/home/foo"

# 测试用例 3：包含 . 和 ..
print(f"测试 4: {sol.simplifyPath('/a/./b/../../c/')}")     # 预期: "/c"

# 测试用例 4：复杂路径
print(f"测试 5: {sol.simplifyPath('/a/../../b/../c//.//')}") # 预期: "/c"
print(f"测试 6: {sol.simplifyPath('/a//b////c/d//././/..')}") # 预期: "/a/b/c"

# 测试用例 5：特殊目录名（... 和 .... 是合法文件名）
print(f"测试 7: {sol.simplifyPath('/.../a/../b/c/.d')}")    # 预期: "/.../b/c/.d"
print(f"测试 8: {sol.simplifyPath('/....')}")               # 预期: "/...."

# 测试用例 6：根目录
print(f"测试 9: {sol.simplifyPath('/')}")                   # 预期: "/"

# 测试用例 7：多级 .. 超出根目录
print(f"测试 10: {sol.simplifyPath('/../../../')}")         # 预期: "/"
````
