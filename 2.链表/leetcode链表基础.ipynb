基于「链表基础.md」设计的循序渐进练习，从遍历到综合题。

````python
from typing import List, Optional

class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next

def list_to_nodes(arr: List[int]) -> Optional[ListNode]:
    """数组 → 链表，返回 head"""
    dummy = ListNode(0)
    cur = dummy
    for v in arr:
        cur.next = ListNode(v)
        cur = cur.next
    return dummy.next

def print_list(head: Optional[ListNode]) -> None:
    """打印链表"""
    vals = []
    cur = head
    while cur:
        vals.append(str(cur.val))
        cur = cur.next
    print(' -> '.join(vals) + ' -> None')

head = list_to_nodes([1, 2, 3, 4, 5])
print('链表创建完成：', end='')
print_list(head)
````

````python
# TODO 2：补全代码，查找链表中是否有值等于 target 的节点
# head = 1 -> 2 -> 3 -> 4 -> 5 -> None
# target = 3  → 返回 True
# target = 6  → 返回 False

def contains(head: Optional[ListNode], target: int) -> bool:
    # 请手写遍历查找
    cur = head
    while cur:
        if cur.val == target:
            return True
        cur = cur.next
    return False

print(contains(head, 3))  # True
print(contains(head, 6))  # False
````

## Part 2：基础操作（删除 & 插入）

**删除当前节点后面的节点：**
```python
cur.next = cur.next.next
```

**插入新节点（顺序不能反）：**
```python
node.next = cur.next
cur.next = node
```

````python
# TODO 3：删除链表中所有值为 val 的节点
# 输入：1 -> 2 -> 6 -> 3 -> 4 -> 5 -> 6 -> None, val = 6
# 输出：1 -> 2 -> 3 -> 4 -> 5 -> None
# 提示：使用 dummy 节点简化头节点删除

def remove_elements(head: Optional[ListNode], val: int) -> Optional[ListNode]:
    dummy = ListNode(0)
    dummy.next = head
    cur = dummy
    while cur.next:
        if cur.next.val == val:
            cur.next = cur.next.next
        else:
            cur = cur.next
    return dummy.next

head2 = list_to_nodes([1, 2, 6, 3, 4, 5, 6])
print('原链表：', end=''); print_list(head2)
head2 = remove_elements(head2, 6)
print('删除后：', end=''); print_list(head2)
````

````python
# TODO 4：在有序链表中插入一个新节点，保持有序
# 输入：1 -> 3 -> 5 -> 7 -> None, 插入 4
# 输出：1 -> 3 -> 4 -> 5 -> 7 -> None

def insert_sorted(head: Optional[ListNode], val: int) -> Optional[ListNode]:
    # 请手写：找到插入位置 → 插入新节点
    dummy = ListNode(0)
    dummy.next = head
    cur = dummy
    while cur.next and cur.next.val < val:
        cur = cur.next
    node = ListNode(val)
    node.next = cur.next
    cur.next = node
    return dummy.next

head3 = list_to_nodes([1, 3, 5, 7])
print('原链表：', end=''); print_list(head3)
head3 = insert_sorted(head3, 4)
print('插入后：', end=''); print_list(head3)
````

````python
# TODO 5：删除链表倒数第 N 个节点（必须用 dummy）
# 输入：1 -> 2 -> 3 -> 4 -> 5 -> None, n = 2
# 输出：1 -> 2 -> 3 -> 5 -> None

def remove_nth_from_end(head: Optional[ListNode], n: int) -> Optional[ListNode]:
    dummy = ListNode(0)
    dummy.next = head
    slow = fast = dummy
    for _ in range(n):
        fast = fast.next
    
    while fast.next:
        slow = slow.next
        fast = fast.next
    
    slow.next = slow.next.next
    return dummy.next


head4 = list_to_nodes([1, 2, 3, 4, 5])
print('原链表：', end=''); print_list(head4)
head4 = remove_nth_from_end(head4, 2)
print('删除后：', end=''); print_list(head4)
````

## Part 4：反转链表（核心基本功）

**四步口诀：先存后 → 再反指 → 前进 prev → 前进 cur**

```python
prev = None
cur = head
while cur:
    nxt = cur.next
    cur.next = prev
    prev = cur
    cur = nxt
return prev
```

````python
# 完整模板演示
def reverse_list(head: Optional[ListNode]) -> Optional[ListNode]:
    prev = None
    cur = head
    while cur:
        nxt = cur.next
        cur.next = prev
        prev = cur
        cur = nxt
    return prev

h = list_to_nodes([1, 2, 3, 4, 5])
print('原链表：', end=''); print_list(h)
h = reverse_list(h)
print('反转后：', end=''); print_list(h)
````

````python
# TODO 6：反转链表从位置 left 到 right 的部分
# 输入：1 -> 2 -> 3 -> 4 -> 5 -> None, left=2, right=4
# 输出：1 -> 4 -> 3 -> 2 -> 5 -> None
# 提示：先找到 left 前一个节点，再反转中间段

def reverse_between(head: Optional[ListNode], left: int, right: int) -> Optional[ListNode]:
    # prev 固定不动

    # cur 固定在反转区间第一个节点

    # 不断把 cur.next
    # 摘下来
    # 插到 prev 后面
    dummy = ListNode(0, head)
    prev = dummy

    for _ in range(left-1):
        prev = prev.next
    cur = prev.next

    for _ in range(right-left):
        nxt = cur.next
        cur.next = nxt.next
        nxt.next = prev.next
        prev.next = nxt
    
    return dummy.next


h = list_to_nodes([1, 2, 3, 4, 5])
print('原链表：', end=''); print_list(h)
h = reverse_between(h, 2, 4)
print('局部反转：', end=''); print_list(h)
````

````python
# TODO 7：找链表的中间节点
# 奇数个节点返回中间，偶数个返回第二个中间
# 输入：1 -> 2 -> 3 -> 4 -> 5 -> None
# 输出：3
# 输入：1 -> 2 -> 3 -> 4 -> None
# 输出：3

def middle_node(head: Optional[ListNode]) -> Optional[ListNode]:
    slow = fast = head
    while fast and fast.next:
        slow = slow.next
        fast = fast.next.next
    return slow

h = list_to_nodes([1, 2, 3, 4, 5])
print('中点值：', middle_node(h).val)  # 3
h = list_to_nodes([1, 2, 3, 4])
print('中点值：', middle_node(h).val)  # 3
````

````python
# TODO 8：判断链表是否有环
# 如果有环返回 True，否则 False
# 提示：快慢指针，fast 追上 slow 则有环

def has_cycle(head: Optional[ListNode]) -> bool:
    slow = fast = head
    while fast and fast.next:
        slow = slow.next
        fast = fast.next.next
        if slow is fast:
            return True
    return False

# 构造有环链表：1 -> 2 -> 3 -> 4 -> 5 -> 3（指向 3）
nodes = [ListNode(i) for i in range(1, 6)]
for i in range(4):
    nodes[i].next = nodes[i+1]
nodes[4].next = nodes[2]  # 成环
print('有环？', has_cycle(nodes[0]))  # True

h = list_to_nodes([1, 2, 3, 4, 5])
print('有环？', has_cycle(h))  # False
````

## Part 6：合并两个有序链表

**模板：**
```python
dummy = ListNode(0)
cur = dummy
while l1 and l2:
    if l1.val <= l2.val:
        cur.next = l1; l1 = l1.next
    else:
        cur.next = l2; l2 = l2.next
    cur = cur.next
cur.next = l1 if l1 else l2
return dummy.next
```

````python
# TODO 9：合并两个有序链表
# 输入：1 -> 2 -> 4 -> None,  1 -> 3 -> 4 -> None
# 输出：1 -> 1 -> 2 -> 3 -> 4 -> 4 -> None

def merge_two_lists(l1: Optional[ListNode], l2: Optional[ListNode]) -> Optional[ListNode]:
    dummy = ListNode(0)
    cur = dummy
    while l1 and l2:
        if l1.val <= l2.val:
            cur.next = l1
            l1 = l1.next
        else:
            cur.next = l2
            l2 = l2.next
        cur = cur.next
    
    cur.next = l1 if l1 else l2
    return dummy.next

l1 = list_to_nodes([1, 2, 4])
l2 = list_to_nodes([1, 3, 4])
merged = merge_two_lists(l1, l2)
print('合并结果：', end=''); print_list(merged)
````

## Part 7：综合挑战

综合运用前面学到的技巧。

````python
# TODO 10：判断回文链表
# 输入：1 -> 2 -> 2 -> 1 -> None  → True
# 输入：1 -> 2 -> None            → False
# 思路：找中点 + 反转后半段 + 比较

def is_palindrome(head: Optional[ListNode]) -> bool:
    if not head:
        return True
    slow = fast = head
    while fast and fast.next:
        slow = slow.next
        fast = fast.next.next
    prev = None
    while slow:
        nxt = slow.next
        slow.next = prev
        prev = slow
        slow = nxt
    left, right = head, prev
    while right:
        if left.val != right.val:
            return False
        left = left.next
        right = right.next
    return True

h = list_to_nodes([1, 2, 2, 1])
print('回文？', is_palindrome(h))  # True
h = list_to_nodes([1, 2])
print('回文？', is_palindrome(h))  # False
````

````python
# TODO 11：两两交换链表中的相邻节点
# 输入：1 -> 2 -> 3 -> 4 -> None
# 输出：2 -> 1 -> 4 -> 3 -> None
# 提示：使用 dummy + 三个指针（prev, a, b）

def swap_pairs(head: Optional[ListNode]) -> Optional[ListNode]:
    dummy = ListNode(0, head)
    prev = dummy
    while prev.next and prev.next.next:
        a = prev.next
        b = a.next
        a.next = b.next
        b.next = a
        prev.next = b
        prev = a
    return dummy.next

h = list_to_nodes([1, 2, 3, 4])
print('原链表：', end=''); print_list(h)
h = swap_pairs(h)
print('交换后：', end=''); print_list(h)
````

````python
# TODO 12：两数相加（最高难度挑战）
# 输入：l1 = 2 -> 4 -> 3 (342), l2 = 5 -> 6 -> 4 (465)
# 输出：7 -> 0 -> 8 (342 + 465 = 807)
# 节点按逆序存储，每位存一个数字

def add_two_numbers(l1: Optional[ListNode], l2: Optional[ListNode]) -> Optional[ListNode]:
    dummy = ListNode(0)
    cur = dummy
    carry = 0
    while l1 or l2 or carry:
        v1 = l1.val if l1 else 0
        v2 = l2.val if l2 else 0
        s = v1 + v2 + carry
        carry = s // 10
        cur.next = ListNode(s % 10)
        cur = cur.next
        if l1: l1 = l1.next
        if l2: l2 = l2.next
    return dummy.next

l1 = list_to_nodes([2, 4, 3])
l2 = list_to_nodes([5, 6, 4])
res = add_two_numbers(l1, l2)
print('相加结果：', end=''); print_list(res)  # 7 -> 0 -> 8
````
