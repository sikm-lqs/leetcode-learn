## 1. 相交链表

- **难度**：简单
- **标签**：链表 / 双指针

**题目**：给定两个单链表的头节点 `headA` 和 `headB`，找出并返回两个单链表相交的起始节点。如果两个链表不存在相交节点，返回 `null`。

**示例**：
```python
输入：intersectVal = 8, listA = [4,1,8,4,5], listB = [5,6,1,8,4,5], skipA = 2, skipB = 3
输出：Intersected at '1'
解释：相交节点的值为 1
```

**思路**：双指针法。指针 `pa` 从链表 A 出发，指针 `pb` 从链表 B 出发。当 `pa` 走到链表 A 末尾时，转向链表 B；当 `pb` 走到链表 B 末尾时，转向链表 A。若两链表相交，`pa` 和 `pb` 会在相交点相遇；否则会同时到达 `None`。路径长度：`pa` 走 `a+c+b`，`pb` 走 `b+c+a`，最终对齐。时间 O(a+b)，空间 O(1)。

**亮点**：通过「切换链表」的方式让两个指针走过相同的路径长度，巧妙地对齐了起点差异；无需计算长度，代码简洁优雅。

````python
# Definition for singly-linked list.
class ListNode:
    def __init__(self, x):
        self.val = x
        self.next = None

class Solution:
    def getIntersectionNode(self, headA: ListNode, headB: ListNode) -> ListNode:
        """
        双指针法：pa 走路径 a+c+b，pb 走路径 b+c+a，在交点相遇
        """
        pa, pb = headA, headB

        while pa != pb:
            pa = pa.next if pa else headB
            pb = pb.next if pb else headA
        
        return pa
````

## 2. 反转链表

- **难度**：简单
- **标签**：链表 / 递归

**题目**：给定单链表的头节点 `head`，请反转链表，并返回反转后的链表。

**示例**：
```python
输入：head = [1, 2, 3, 4, 5]
输出：[5, 4, 3, 2, 1]
```

**思路**：迭代法。维护 `prev` 和 `cur` 两个指针。遍历时保存 `cur.next`，然后将 `cur.next` 指向 `prev`，最后移动 `prev` 和 `cur`。时间 O(n)，空间 O(1)。

**亮点**：只需三个变量（`prev`、`cur`、`nxt`）即可完成原地反转，空间 O(1)；关键是保存 `cur.next` 后再修改指针，防止链表断裂。

````python
# Definition for singly-linked list.
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next

class Solution:
    def reverseList(self, head: ListNode) -> ListNode:
        """
        迭代法：原地反转链表
        """
        prev, cur = None, head
        
        while cur:
            nxt = cur.next      # 先保存下一个节点
            cur.next = prev     # 当前节点指向前一个节点（反转）
            prev = cur          # prev 前进
            cur = nxt           # cur 前进
        
        return prev
````

## 3. 回文链表

- **难度**：简单
- **标签**：链表 / 双指针

**题目**：给定单链表的头节点 `head`，判断该链表是否为回文链表。如果是，返回 `true`；否则，返回 `false`。

**示例**：
```python
输入：head = [1, 2, 2, 1]
输出：true

输入：head = [1, 2]
输出：false
```

**思路**：
1. **快慢指针 + 反转**：先用快慢指针找到中点，将后半段链表原地反转，然后逐个比较前后半段。最后可选择恢复后半段。时间 O(n)，空间 O(1)。

2. **数组法**：将链表值复制到数组，用双指针判断回文。时间 O(n)，空间 O(n)。

**亮点**：快慢指针找中点是经典技巧，`slow` 每次走一步，`fast` 每次走两步；反转后半段后只需一次遍历即可完成比较，空间 O(1)。

````python
# Definition for singly-linked list.
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next

class Solution:
    def isPalindrome(self, head: ListNode) -> bool:
        """
        快慢指针 + 反转后半段
        """
        # 找中点
        slow = fast = head
        while fast and fast.next:
            slow = slow.next
            fast = fast.next.next
        
        # 反转后半段（slow 为中点或中点后第一个）
        prev = None
        cur = slow
        while cur:
            nxt = cur.next
            cur.next = prev
            prev = cur
            cur = nxt
        
        # 逐个比较
        left, right = head, prev
        while right:
            if left.val != right.val:
                return False
            left = left.next
            right = right.next
        
        return True
````

## 4. 环形链表

- **难度**：简单
- **标签**：链表 / 双指针

**题目**：给定链表的头节点 `head`，判断链表中是否有环。如果链表中存在某个节点可以通过连续跟踪 `next` 指针再次到达，则存在环。

**示例**：
```python
输入：head = [3, 2, 0, -4], pos = 1（表示尾节点连接到索引 1 的节点）
输出：true
解释：链表中有一个环，其尾部连接到第二个节点
```

**思路**：快慢指针法。`slow` 每次走一步，`fast` 每次走两步。若存在环，`fast` 最终会追上 `slow`（两指针相遇）；若不存在环，`fast` 会先到达 `None`。时间 O(n)，空间 O(1)。

**亮点**： Floyd 判圈算法，又称龟兔赛跑算法；`fast` 和 `slow` 从同一点出发，`fast` 相对于 `slow` 的速度为 1，若有环必相遇；无需额外空间，只用两个指针即可完成。

````python
# Definition for singly-linked list.
class ListNode:
    def __init__(self, x):
        self.val = x
        self.next = None

class Solution:
    def hasCycle(self, head: ListNode) -> bool:
        """
        快慢指针法（Floyd 判圈算法）
        """
        slow = fast = head
        
        while fast and fast.next:
            slow = slow.next
            fast = fast.next.next
            if slow is fast:
                return True
        
        return False
````

## 5. 环形链表 II

- **难度**：中等
- **标签**：链表 / 双指针

**题目**：给定链表的头节点 `head`，返回链表开始入环的第一个节点。如果链表无环，返回 `null`。不允许修改链表。

**示例**：
```python
输入：head = [3, 2, 0, -4], pos = 1
输出：返回索引为 1 的链表节点
解释：链表中有一个环，其尾部连接到第二个节点
```

**思路**：快慢指针法分两步。第一步：`fast` 和 `slow` 从同一点出发，判断是否有环并找到相遇点。第二步：将一个指针放回 `head`，两指针每次都走一步，再次相遇的位置就是入环点。数学原理：设头到入环点距离为 a，入环点到相遇点为 b，相遇点到入环点为 c，则有 2(a+b) = a+b+c+b → a = c。时间 O(n)，空间 O(1)。

**亮点**：利用数学关系 a = c，将「找入环点」转化为第二次相遇问题；第一次相遇判断是否有环，第二次相遇精确定位入环点，无需额外空间。

````python
# Definition for singly-linked list.
class ListNode:
    def __init__(self, x):
        self.val = x
        self.next = None

class Solution:
    def detectCycle(self, head: ListNode) -> ListNode:
        """
        快慢指针法：找入环的第一个节点
        """
        slow = fast = head
        
        # 第一步：判断是否有环并找到相遇点
        while fast and fast.next:
            slow = slow.next
            fast = fast.next.next
            if slow is fast:
                # 第二步：一个指针回到 head，两指针同速前进
                p = head
                while p is not slow:
                    p = p.next
                    slow = slow.next
                return p
        
        return None
````

## 6. 合并两个有序链表

- **难度**：简单
- **标签**：链表 / 递归

**题目**：将两个升序链表合并为一个新的升序链表并返回。新链表是通过拼接给定的两个链表的所有节点组成的。

**示例**：
```python
输入：list1 = [1, 2, 4], list2 = [1, 3, 4]
输出：[1, 1, 2, 3, 4, 4]

输入：list1 = [], list2 = []
输出：[]
```

**思路**：迭代法。使用哑节点 `dummy` 简化头节点处理。用指针 `cur` 遍历，每次比较 `list1` 和 `list2` 当前节点值，将较小的接入结果链表。时间 O(m+n)，空间 O(1)。

**亮点**：哑节点技巧避免了单独处理头节点的特殊情况；只需遍历一次，同时处理比较和接入，代码简洁高效。

````python
# Definition for singly-linked list.
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next

class Solution:
    def mergeTwoLists(self, list1: ListNode, list2: ListNode) -> ListNode:
        """
        迭代法：合并两个有序链表
        """
        dummy = ListNode(0)
        cur = dummy
        
        while list1 and list2:
            if list1.val <= list2.val:
                cur.next = list1
                list1 = list1.next
            else:
                cur.next = list2
                list2 = list2.next
            cur = cur.next
        
        # 接上剩余的链表
        cur.next = list1 if list1 else list2
        
        return dummy.next
````

## 7. 两数相加

- **难度**：中等
- **标签**：链表 / 数学

**题目**：给你两个非空的链表，表示两个非负的整数。每位数字都是按照逆序的方式存储，每个节点只能存储一位数字。将两个数相加，并以相同形式返回一个表示和的链表。

**示例**：
```python
输入：l1 = [2, 4, 3], l2 = [5, 6, 4]
输出：[7, 0, 8]
解释：342 + 465 = 807

输入：l1 = [9, 9, 9], l2 = [1]
输出：[0, 0, 0, 1]
解释：999 + 1 = 1000
```

**思路**：模拟加法。同时遍历两个链表，对相同位置的数字相加并加上进位。当前位结果为 `(x + y + carry) % 10`，新的进位为 `(x + y + carry) // 10`。当两个链表都遍历完且进位为 0 时结束。时间 O(max(m, n))，空间 O(max(m, n))。

**亮点**：逆序存储恰好对应从低位到高位的加法顺序；使用哑节点简化结果链表的头节点处理；处理完所有节点后还需检查剩余进位。

````python
# Definition for singly-linked list.
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next

class Solution:
    def addTwoNumbers(self, l1: ListNode, l2: ListNode) -> ListNode:
        """
        模拟加法：逆序存储，逐位相加并处理进位
        """
        dummy = cur = ListNode(0)
        carry = 0

        while l1 or l2 or carry:
            # 取出当前位的值（如果链表已空则为 0）
            x = l1.val if l1 else 0
            y = l2.val if l2 else 0

            # 计算和与进位
            s = x + y + carry
            carry = s // 10
            cur.next = ListNode(s % 10)
            cur = cur.next

            # 移动到下一位
            if l1:
                l1 = l1.next
            if l2:
                l2 = l2.next
        
        return dummy.next
````

## 8. 删除链表的倒数第 N 个结点

- **难度**：中等
- **标签**：链表 / 双指针

**题目**：给定一个链表，删除链表的倒数第 n 个结点，并且返回链表的头结点。

**示例**：
```python
输入：head = [1, 2, 3, 4, 5], n = 2
输出：[1, 2, 3, 5]

输入：head = [1], n = 1
输出：[]
```

**思路**：双指针法。使用哑节点 `dummy` 指向头节点（防止删除第一个节点）。`fast` 先走 n+1 步，然后 `slow` 和 `fast` 同时前进，当 `fast` 到达 `None` 时，`slow` 的下一个节点就是要删除的节点。时间 O(n)，空间 O(1)。

**亮点**：哑节点技巧消除了删除头节点的特殊情况；`fast` 先走 n+1 步确保 `slow` 停在被删节点的前一个节点，便于删除操作。

````python
# Definition for singly-linked list.
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next

class Solution:
    def removeNthFromEnd(self, head: ListNode, n: int) -> ListNode:
        """
        双指针法：删除倒数第 n 个结点
        """
        # 哑节点：防止删除第一个节点
        dummy = ListNode(0, head)
        
        slow = fast = dummy
        
        # fast 先走 n+1 步
        for _ in range(n + 1):
            fast = fast.next
        
        # slow 和 fast 同时前进，直到 fast 到达末尾
        while fast:
            slow = slow.next
            fast = fast.next
        
        # 删除 slow 的下一个节点
        slow.next = slow.next.next
        
        return dummy.next
````

## 9. 两两交换链表中的节点

- **难度**：中等
- **标签**：链表 / 递归

**题目**：给定一个链表，两两交换其中相邻的节点，并返回交换后链表的头节点。必须在不修改节点内部的值的情况下完成本题（即只能进行节点交换）。

**示例**：
```python
输入：head = [1, 2, 3, 4]
输出：[2, 1, 4, 3]

输入：head = []
输出：[]

输入：head = [1]
输出：[1]
```

**思路**：迭代法。使用哑节点 `dummy` 简化头节点处理。每次取出相邻的两个节点 `first` 和 `second`，记录剩余链表 `rest`，交换 `first` 和 `second` 的位置，然后移动 `cur` 指针到 `first`（交换后的第二个节点）。时间 O(n)，空间 O(1)。

**亮点**：三指针交换法清晰地完成相邻节点的两两交换；哑节点技巧统一了偶数和奇数长度链表的处理逻辑；关键在于正确保存后续链表 `rest` 防止断链。

````python
# Definition for singly-linked list.
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next

class Solution:
    def swapPairs(self, head: ListNode) -> ListNode:
        """
        迭代法：两两交换相邻节点
        """
        dummy = ListNode(0, head)
        cur = dummy
        
        while cur.next and cur.next.next:
            # 保存两个节点和后续链表
            first = cur.next
            second = cur.next.next
            rest = second.next
            
            # 交换两个节点
            cur.next = second
            second.next = first
            first.next = rest
            
            # 移动到下一对的第二个节点
            cur = first
        
        return dummy.next
````

## 10. 随机链表的复制

- **难度**：中等
- **标签**：链表 / 哈希表

**题目**：给定一个长度为 n 的链表，每个节点包含一个额外增加的随机指针 `random`，该指针可以指向链表中的任何节点或空节点。构造这个链表的深拷贝。

**示例**：
```python
输入：链表中有节点 [7,null,13,0,11,4,15,null]，random 指针关系为：7->13, 13->11, 11->4, 4->15, 15->0
输出：深拷贝后的链表，random 指针关系与原链表一致
```

**思路**：哈希表 + 两次遍历。第一遍遍历：复制所有节点，用哈希表记录 `原节点 → 新节点` 的映射。第二遍遍历：根据哈希表补全每个节点的 `next` 和 `random` 指针。时间 O(n)，空间 O(n)。

**亮点**：哈希表建立了原节点和新节点的映射关系，使得第二次遍历时能快速找到新节点；分离复制值和连接指针的过程，逻辑清晰，避免了指针混乱。

````python
# Definition for a Node.
class Node:
    def __init__(self, x: int, next: 'Node' = None, random: 'Node' = None):
        self.val = int(x)
        self.next = next
        self.random = random

class Solution:
    def copyRandomList(self, head: 'Node') -> 'Node':
        """
        哈希表 + 两次遍历
        """
        if not head:
            return None
        
        old_to_new = {}
        cur = head
        
        # 第一遍：复制所有节点
        while cur:
            old_to_new[cur] = Node(cur.val)
            cur = cur.next
        
        # 第二遍：补上 next 和 random
        cur = head
        while cur:
            old_to_new[cur].next = old_to_new.get(cur.next)
            old_to_new[cur].random = old_to_new.get(cur.random)
            cur = cur.next
        
        return old_to_new[head]
````

## 11. 排序链表

- **难度**：中等
- **标签**：链表 / 排序

**题目**：给定链表的头结点 `head`，请将其按升序排序并返回排序后的链表。

**示例**：
```python
输入：head = [4, 2, 1, 3]
输出：[1, 2, 3, 4]

输入：head = [-1, 5, 3, 4, 0]
输出：[-1, 0, 3, 4, 5]
```

**思路**：链表 → 数组 → 排序 → 链表。遍历链表将所有值存入数组，对数组排序，最后根据排序后的数组重建链表。时间 O(n log n)，空间 O(n)。

**亮点**：将链表排序问题转化为数组排序，利用 Python 高效的 Timsort；虽然需要 O(n) 额外空间，但代码简洁易懂，适合面试场景。

````python
# Definition for singly-linked list.
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next

class Solution:
    def sortList(self, head: ListNode) -> ListNode:
        """
        链表 → 数组 → 排序 → 链表
        """
        # 链表转数组
        nums = []
        cur = head
        while cur:
            nums.append(cur.val)
            cur = cur.next

        # 排序
        nums.sort()

        # 数组转链表
        dummy = ListNode(0)
        cur = dummy
        for num in nums:
            cur.next = ListNode(num)
            cur = cur.next
        
        return dummy.next
````

## 12. LRU 缓存

- **难度**：中等
- **标签**：设计 / 哈希表 / 链表

**题目**：设计并实现满足 LRU (最近最少使用) 缓存约束的数据结构。实现 `LRUCache` 类：
- `LRUCache(int capacity)`：以正整数作为容量 `capacity` 初始化 LRU 缓存
- `int get(int key)`：如果关键字 `key` 存在于缓存中，返回关键字的值；否则返回 -1
- `void put(int key, int value)``：如果关键字 `key` 已经存在，变更其数据值；如果不存在，插入该组 `key-value`。如果插入操作导致关键字数量超过 `capacity`，则逐出最久未使用的关键字。

**示例**：
```python
输入
["LRUCache", "capacity", "put", "get", "put", "get", "get", "put", "get", "get"]
[[2], [1, 1], [1], [2, 1], [4], [1, 1], [1], [2], [4], [4]]

输出
[null, null, null, 1, null, -1, null, -1, 3, 4]
解释
LRUCache 缓存容量为 2
put(1, 1)：缓存 = {1=1}
get(1)：返回 1
put(2, 2)：缓存 = {1=1, 2=2}（容量已满）
get(1)：返回 1（key=1 变为最近使用）
put(3, 3)：缓存 = {2=2, 3=3}（逐出 key=1）
get(2)：返回 -1（key=2 已被逐出）
```

**思路**：哈希表 + 双向链表。哈希表提供 O(1) 的查找，双向链表维护访问顺序。最近使用的节点移到链表头部，最久未使用的在尾部。`get` 操作将节点移到头部，`put` 操作插入新节点到头部，容量超限时删除尾部节点并同步删除哈希表中的键。时间 O(1)，空间 O(capacity)。

**亮点**：哈希表提供 O(1) 查找，双向链表维护顺序；双向链表的 `remove` 和 `add_to_head` 操作都是 O(1)；将访问和插入统一为「移到头部」的操作，逻辑简洁高效。

````python
class Node:
    def __init__(self, key=0, value=0):
        self.key = key
        self.value = value
        self.prev = None
        self.next = None

class LRUCache:
    def __init__(self, capacity: int):
        self.capacity = capacity
        self.map = {}
        
        # 哨节点和尾节点
        self.head = Node()
        self.tail = Node()
        self.head.next = self.tail
        self.tail.prev = self.head

    def get(self, key: int) -> int:
        if key not in self.map:
            return -1
        
        node = self.map[key]
        # 访问后移到头部（变成最近使用）
        self._remove(node)
        self._add_to_head(node)
        return node.value

    def put(self, key: int, value: int) -> None:
        if key in self.map:
            # key 存在：更新值并移到头部
            node = self.map[key]
            node.value = value
            self._remove(node)
            self._add_to_head(node)
        else:
            # key 不存在：创建新节点并插入头部
            node = Node(key, value)
            self.map[key] = node
            self._add_to_head(node)
            
            # 超出容量：删除尾部节点
            if len(self.map) > self.capacity:
                old = self.tail.prev
                self._remove(old)
                del self.map[old.key]

    def _remove(self, node: Node) -> None:
        """从双向链表中删除节点"""
        node.prev.next = node.next
        node.next.prev = node.prev

    def _add_to_head(self, node: Node) -> None:
        """将节点添加到双向链表头部"""
        node.next = self.head.next
        node.prev = self.head
        self.head.next.prev = node
        self.head.next = node
````
